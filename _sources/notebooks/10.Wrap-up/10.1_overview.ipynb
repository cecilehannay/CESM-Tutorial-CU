## What we didn't cover

Just a few examples of topics beyond the scope of this tutorial:

- `SourceMods`: modifying the model source code itself.
- Adding new output fields to a component.
- Configuring individual component (CAM, CLM, POP, CICE, ...) science options.
- Running single-component (data-driven) configurations.
- Post-processing and diagnostics packages (e.g., the CESM diagnostics suite, ADF).
- Setting up and running on other HPC systems / the cloud.

The full [CESM Tutorial](https://ncar.github.io/CESM-Tutorial/) website and the
[CESM webpage](http://www.cesm.ucar.edu/models/cesm2/) are the best starting points for all
of these.
