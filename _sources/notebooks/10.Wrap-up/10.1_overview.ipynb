# Tutorial website and further resources

## The full CESM Tutorial

A much more in-depth, one-week CESM tutorial (lectures as well as hands-on material) is offered
periodically at NCAR. The materials are online, and you can apply to attend:
http://www.cesm.ucar.edu/events/tutorials/

![CESM tutorial website](../../images/tutorial_website_2024.png)

## What we didn't cover

A few examples of topics beyond the scope of this tutorial:

- `SourceMods`: modifying the model source code itself.
- Adding new output fields to a component.
- Configuring the science options of individual components (CAM, CLM, POP, CICE, ...).
- Running single-component (data-driven) configurations.
- Post-processing and diagnostics packages (for example, the CESM diagnostics suite and ADF).
- Setting up and running CESM on other HPC systems or in the cloud.

The full [CESM Tutorial](https://ncar.github.io/CESM-Tutorial/) website and the
[CESM webpage](http://www.cesm.ucar.edu/models/cesm2/) are the best starting points for all of
these. The [additional materials](../11.Additional-Materials/title.ipynb) in this book also
include a few optional exercises that go further.
