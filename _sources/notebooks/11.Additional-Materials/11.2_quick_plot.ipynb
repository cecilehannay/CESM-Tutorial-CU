## Open your output file

Point this at one of your own monthly history files — for example, the one from the
[run-length homework](../05.Run-length/5.4_homework.ipynb) or [Exercise 2](../05.Run-length/5.3_exercise2.ipynb):


````python
import xarray as xr

# update this path to point at one of your own history files
path = "/glade/derecho/scratch/$USER/archive/b1850_hw/atm/hist/b1850_hw.cam.h0.0001-01.nc"

ds = xr.open_dataset(path)
ds
````

Running the cell above prints a summary of every variable, dimension, and piece of metadata
in the file — a nice way to see what's actually in a CESM history file without `ncdump -h`.


## Make a quick map

`xarray` variables have a built-in `.plot()` that makes a quick map (or line plot, for a
1-D variable) with almost no code. Let's plot surface air temperature (`TREFHT`):


````python
import matplotlib.pyplot as plt

ds["TREFHT"].isel(time=0).plot()
plt.title("Surface air temperature (TREFHT)")
plt.show()
````

## Try it yourself

- Pick a different variable from the list printed above (e.g. `PRECT` for precipitation) and
  plot it the same way.
- Instead of a map, try a global-mean time series:
  ```python
  ds["TREFHT"].mean(["lat", "lon"]).plot()
  ```
- If your case has more than one monthly file, open them all at once with
  `xr.open_mfdataset("/path/to/*.nc")` instead of `xr.open_dataset(...)`.
