# Additional Materials

This part is optional, self-paced material — work through it if there's time in class, or
afterward on your own. Nothing here is required to finish the core tutorial.

In this part:
- Why you might want to use CESM, and where to find CESM simulations that are already
  freely available if you don't need to run your own.
- A quick introduction to looking at your own CESM output with Python instead of `ncview`.
- The namelist "challenge" exercises from the [Namelist](../07.Namelist/title.ipynb) chapter
  (high-frequency output, a CAM tuning parameter, prescribed SST).
