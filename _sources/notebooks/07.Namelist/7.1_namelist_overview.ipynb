# CESM namelists

A **namelist** is a plain text file of settings that a model component reads when it starts.
Each component has its own: `atm_in` for the atmosphere, `lnd_in` for the land, `ocn_in` or
`pop_in` for the ocean, `ice_in` for sea ice, and so on. A namelist contains variables such as:

- `co2vmr`: the CO2 concentration
- `dust_emis_fact`: a tuning factor for dust emissions
- `bnd_topo`: the name of the topography input file

You can see the namelists of your case in the run directory. They are the files whose names
end in `_in`: `atm_in`, `cism_in`, `drv_flds_in`, `drv_in`, `ice_in`, `lnd_in`, `mosart_in`,
`pop_in`, `wav_in`.

## How to change a namelist

You never edit the `*_in` files directly. CESM regenerates them, so your edits would be lost.
Instead:

1. Put your changes in the `user_nl_*` files in the **case directory** (`user_nl_cam` for the
   atmosphere, `user_nl_clm` for the land, `user_nl_pop` for the ocean, ...).
2. CESM reads the `user_nl_*` files and generates the real namelists. This happens
   automatically during `case.build` and `case.submit`, or when you run `./preview_namelists`
   in the case directory.
3. The generated namelists are written to two places: `CaseDocs/` in the case directory, and
   the run directory, where the model reads them.

```text
~/cases/case01                                (case directory)
    user_nl_cam, user_nl_cice, user_nl_cism, user_nl_clm,
    user_nl_cpl, user_nl_mosart, user_nl_pop, user_nl_ww3    <- you edit these
    CaseDocs/
        atm_in, cism_in, drv_flds_in, drv_in, ice_in,
        lnd_in, mosart_in, pop_in, wav_in, ...                <- generated copy, for reference

/glade/derecho/scratch/$USER/case01            (build/run directory)
    bld/
    run/
        atm_in, cism_in, drv_flds_in, drv_in, ice_in,
        lnd_in, mosart_in, pop_in, wav_in, ...                <- generated, read by the model
```

:::{note}
`CaseDocs/` is a copy for reference only. It is handy for checking that your change was picked
up, but the model reads the namelists in the run directory.
:::
