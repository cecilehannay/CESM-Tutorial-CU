# Exercise 5: Double the CO2 concentration

<div class="alert alert-info">

<strong>Exercise 5</strong>

Create a new case `B1850_2x` for the `B1850` compset at `f19_g17`. Double the CO2
concentration.

Check that the value of `co2vmr` changed in `CaseDocs/atm_in` and in the run directory. You
don't need to build (`qcmd -- ./case.build`) or submit (`./case.submit`).

</div>


<div class="alert alert-warning">
<details>
<summary><font face="Times New Roman" color='blue'>Click here for hints</font></summary><br>

```bash
cd ~/code/release-cesm2.1.5/cime/scripts
./create_newcase --case ~/cases/B1850_2x --compset B1850 --res f19_g17
cd ~/cases/B1850_2x
./case.setup
```

Find the default value of `co2vmr` (look in `CaseDocs/atm_in` after `case.setup`), then add
double that value to `user_nl_cam`:
```
&cam_inparm
co2vmr = <2x the default value>
/
```

Regenerate the namelists and check the value changed:
```bash
./preview_namelists
grep co2vmr CaseDocs/atm_in
```
