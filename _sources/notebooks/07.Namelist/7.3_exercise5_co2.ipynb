# Exercise 5: Double the CO2 concentration

In this exercise, you create a new case and change one namelist variable. You do not need to
build or submit the case; the goal is to see your change appear in the generated namelist.

:::{admonition} Exercise 5: Double the CO2 concentration
:class: exercise
1. Create a new case named `B1850_2x` with the `B1850` compset at `f19_g17` resolution.
2. Find the default value of `co2vmr` and double it.
3. Check that the new value appears in `CaseDocs/atm_in` and in the run directory.
:::

If Derecho does not recognize your project code when you create the case, see
[Project code issues](../09.Troubleshooting/9.2_project_code_issues.ipynb).

:::{admonition} Hints
:class: hint dropdown
- Create and set up the case exactly as in Exercise 1, just with a different case name.
- After `./case.setup`, the generated namelists are in `CaseDocs/`. Use `grep` to find the
  current value of `co2vmr` in `CaseDocs/atm_in`.
- Put the new value in `user_nl_cam`, inside a `&cam_inparm ... /` block.
- Run `./preview_namelists` to regenerate the namelists, then `grep` again.
:::

:::{admonition} Solution
:class: solution dropdown
Create and set up the case:

```bash
cd ~/code/release-cesm2.1.5/cime/scripts
./create_newcase --case ~/cases/B1850_2x --compset B1850 --res f19_g17
cd ~/cases/B1850_2x
./case.setup
```

Find the default value:

```bash
./preview_namelists
grep co2vmr CaseDocs/atm_in
```

For B1850 you should find a value close to `284.7e-6` (284.7 ppm). Doubling it gives
`569.4e-6`. Add this to `user_nl_cam`:

```fortran
&cam_inparm
co2vmr = 569.4e-6
/
```

Regenerate the namelists and check:

```bash
./preview_namelists
grep co2vmr CaseDocs/atm_in
grep co2vmr /glade/derecho/scratch/$USER/B1850_2x/run/atm_in
```
:::

## How to check it worked

Both `grep` commands in the solution should print your new value, `co2vmr = 569.4e-6`. If you
still see the default, make sure you ran `./preview_namelists` after editing `user_nl_cam`.
