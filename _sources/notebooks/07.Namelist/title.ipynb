# Namelist Modifications

This is where you learn how to change model settings that don't require recompiling the
code — for instance, changing the CO2 concentration or changing CESM input files (such as
topography).

<div class="alert alert-info">
Note: the prescribed SST is changed differently, via <code>xmlchange</code> of
<code>SSTICE_DATA_FILENAME</code> — see below.
</div>
