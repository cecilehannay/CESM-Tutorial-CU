# Run types: Startup, Branch, Hybrid

Every CESM run has to start from *somewhere*: either from nothing (a cold start), or from the
saved state of a previous run (a restart). The three run types below are really just three
different answers to the same question: "where do I start from, and what am I allowed to
change along the way?"

- **Startup** is the ordinary way to run CESM. The very first segment of a run is a *cold
  start* — no restart files, everything begins from scratch. Every later segment (once you set
  `CONTINUE_RUN = TRUE`) picks up exactly where the last one left off, with nothing else
  allowed to change. This is what you've been doing all along in the [Run-length](../05.Run-length/title.ipynb)
  exercises.
- **Branch** starts from another run's restart files, but stays bit-for-bit identical to that
  run — you can add extra output, but you can't change the science or the start date. Think of
  it as "continuing someone else's run, unmodified."
- **Hybrid** also starts from another run's restart files, but this time you *are* allowed to
  change namelists, apply source modifications, and pick a new start date. This is the one
  you'd reach for to kick off a new experiment from an already spun-up state — for example,
  branching a CO2-quadrupling experiment off of a pre-industrial control run partway through.

| Run type | Initialization | Namelist / SourceMods changes | Start date | Bit-for-bit with control? |
|---|---|---|---|---|
| **Startup: control** | Cold start (no restart files) | — | — | — |
| **Startup: `CONTINUE_RUN=TRUE`** | Restart files of the same run | None allowed (only extra output) | Cannot change | Yes |
| **Branch** | Restart files of a previous CESM run | No source mods or namelist changes (except extra output) | Cannot change | Yes |
| **Hybrid** | Restart files of a previous CESM run | Namelist changes and source modifications allowed | Can use a different start date | No |

<div class="alert alert-info">
CESM often uses the <strong>hybrid</strong> run type to start a new experiment from a
spun-up state (e.g., branching a CO2-quadrupling experiment off of a pre-industrial control
run).
</div>

Don't worry about memorizing this table. The one thing worth taking away is: **startup and
branch stay bit-for-bit identical to their source run; hybrid is the only one that lets you
actually change something.**

## The relevant `env_run.xml` variables

| Run type | Key variables |
|---|---|
| Startup (control) | `RUN_TYPE = STARTUP` |
| Startup, continuing | `RUN_TYPE = STARTUP`, `CONTINUE_RUN = TRUE` |
| Branch | `RUN_TYPE = BRANCH`, `RUN_REFCASE`, `RUN_REFDATE` |
| Hybrid | `RUN_TYPE = HYBRID`, `RUN_REFCASE`, `RUN_REFDATE`, `RUN_STARTDATE` |
