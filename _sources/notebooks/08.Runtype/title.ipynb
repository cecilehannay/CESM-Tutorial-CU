# Types of runs: startup, branch, and hybrid

In this section, we look at the different ways a CESM run can be initialized: **startup**,
**branch**, and **hybrid**.

This is one of the more abstract topics in the tutorial — the three run types differ in subtle
ways, and it's easy to mix them up at first. Don't worry if it takes a second read; the
hands-on exercise later in this chapter is optional, so you can always come back to it once
the ideas have had time to sink in.
