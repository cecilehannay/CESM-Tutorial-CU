# Restart files

To continue a run, CESM needs to know exactly what the model looked like when it stopped. That information is saved in **restart files**.

## What a restart file is

A restart file is a snapshot of the complete state of the model (the atmosphere, ocean, land, sea ice, and so on) at one instant in time. With these files, CESM can pick up a run exactly where it stopped, as if it had never been interrupted.

CESM writes a set of restart files, one per component, at the end of each run segment. For a 5-day run of `case01` starting on January 1 of year 1, the restart files are dated January 6, year 1 (`0001-01-06`).

## Where the restart files are

- While the run is in progress, the latest restart files are in your **run directory**.
- Once the run has been archived, they are moved to the **archive directory**, under `rest/`.

You will see files such as:

```text
case01.cam.r.0001-01-06-00000.nc
case01.cam.rs.0001-01-06-00000.nc
case01.cice.r.0001-01-06-00000.nc
case01.pop.ro.0001-01-06-00000
case01.ww3.r.0001-01-06-00000
...
rpointer.atm
rpointer.drv
rpointer.ice
rpointer.lnd
rpointer.ocn.restart
rpointer.rof
...
```

The `rpointer.*` files are small text files, one per component, that tell CESM which restart file to use when it continues the run. You do not normally need to edit them.
