# Exercise 2: Extend a run

Your `case01` run from Exercise 1 stopped after 5 days. In this exercise, you will restart it and run it for one more month.

![Timeline of Exercise 2: 5 days, then one more month](../../images/05.exercise2_timeline.png)

| | Initial run (Exercise 1) | Extension (this exercise) |
|---|---|---|
| Simulates | Jan 1 to Jan 5 (5 days) | Jan 5 to Feb 5 (1 month) |
| Starts from | initial conditions | the restart files written on Jan 5 |
| `CONTINUE_RUN` | `FALSE` | `TRUE` |

:::{admonition} Exercise 2: Extend a run
:class: exercise

1. Restart your run `case01` and extend it for one month.
2. Check that the run completed successfully.
:::

## What you need to change

Everything you need is in the [xml files](5.1_xml_files.ipynb) page: you have to tell CESM to continue from where it stopped rather than start over, and to run for one month instead of five days.

:::{note}
When you extend a run, there is no need to run `case.setup` or `case.build` again. The
executable already exists. You only change the settings and submit.
:::

:::{admonition} Hints
:class: hint dropdown

- You need to change three settings in `env_run.xml`: one that tells CESM this is a continuation of an existing run, and two that set the length of the run.
- Run all the commands from inside your case directory, `~/cases/case01`.
- You do not need to build the model again. Submit the run with `./case.submit`.
:::

:::{admonition} Solution
:class: solution dropdown

From your case directory:

```bash
cd ~/cases/case01

./xmlchange CONTINUE_RUN=TRUE
./xmlchange STOP_OPTION=nmonths,STOP_N=1

./case.submit
```
:::
