# Homework: Extend a run

<div class="alert alert-info">

<strong>Homework</strong>

Create a case called `b1850_hw` with the `B1850` compset at `f19_g17` resolution, in the
`cases` subdirectory of your home directory. Make a 2-month run using two consecutive 1-month
submissions (i.e., you want the run to automatically resubmit itself at the end of the first
month).

Explicitly specify the values you used for the variables `STOP_N`, `STOP_OPTION`,
`RESUBMIT`, `CONTINUE_RUN`.

Send the list of commands you used to both Aneesh and Cecile on Slack!

doubt, please ask — our project code has finite shared resources.
