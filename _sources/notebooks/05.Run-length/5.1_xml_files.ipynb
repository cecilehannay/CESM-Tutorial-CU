# The xml files and run length

Every setting that controls how your case runs, including how long it runs, is stored in a handful of **xml files** inside the case directory. This page shows you where those settings live and how to read and change them.

## The `env_*.xml` files

When you ran `create_newcase`, CESM wrote several files named `env_*.xml` into your case directory (for example `~/cases/case01`). Together they describe everything about the experiment: which components it uses, how it is built, and how it runs.

You never edit these files by hand. Instead, from inside the case directory, you **read** a setting with `./xmlquery` and **change** it with `./xmlchange`.

:::{tip}
`xmlquery` = look at a setting. `xmlchange` = change a setting. Both must be run from inside your case directory.
:::

The four files you will see are:

- `env_case.xml`: set by `create_newcase` and cannot be modified afterwards.
- `env_mach_pes.xml`: specifies how the components are laid out across processors.
- `env_build.xml`: specifies how the model is built.
- `env_run.xml`: sets run-time information, such as the length of the run and the number of submissions.

You will interact with `env_run.xml` most frequently.

## How long does the run last? `STOP_N` and `STOP_OPTION`

Two settings in `env_run.xml` control the length of a run: `STOP_N` (how many) and `STOP_OPTION` (of what unit).

By default, a new case is set to run for 5 days:

```text
STOP_N = 5
STOP_OPTION = ndays
```

You can check these values yourself from your case directory:

```bash
cd ~/cases/case01
./xmlquery STOP_OPTION,STOP_N
```

The output looks like:

```text
    STOP_OPTION: ndays
    STOP_N: 5
```

To increase the run length from 5 days to 1 month, change both settings at once:

```bash
./xmlchange STOP_OPTION=nmonths,STOP_N=1
```

Run `./xmlquery STOP_OPTION,STOP_N` again to confirm the change.

## Continue or resubmit a run: `CONTINUE_RUN` and `RESUBMIT`

Two more settings in `env_run.xml` control what happens when a run segment finishes:

- `CONTINUE_RUN`
  - `FALSE`: this is the first segment of your run. The model starts from scratch.
  - `TRUE`: you are continuing a run. The model picks up where the previous segment stopped.
- `RESUBMIT`: if greater than 0, the case automatically resubmits itself that many times once the current segment finishes. Each new segment continues from the previous one.

For example, if you set:

```text
STOP_N = 1
STOP_OPTION = nmonths
RESUBMIT = 3
```

the run proceeds one month at a time and resubmits itself 3 times, for a total of 4 months (the first submission plus 3 resubmissions).
