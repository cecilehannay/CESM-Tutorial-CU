# CESM Workflow: Super Quick Start

You have downloaded the CESM code. Now you are ready to create and run your first CESM experiment, or **case**.

The basic **CESM workflow** uses four main commands:

1. **Create** a new case.
2. **Set up** the case.
3. **Build** the model executable.
4. **Submit** the simulation to Derecho.

Each command performs a different step, but together they take you from the CESM source code to a running simulation. 

By the end of this section, you will have submitted your own CESM case and learned where CESM creates its case, build/run, and archive directories.
