# Project code issues

Every job on Derecho must be charged to a project allocation. Normally CESM picks up your
project code from the `PROJECT` variable you set in `~/.bashrc` during the
[prerequisites](../01.Prerequisites/1.1_login_and_project.ipynb). If something went wrong
there, `create_newcase`, `case.build`, or `case.submit` may complain about a missing or invalid
project. Here is how to fix it for one case.

## Set the project code in the case

From your case directory, check the current values and set them explicitly:

```bash
./xmlquery PROJECT,JOB_QUEUE,JOB_WALLCLOCK_TIME
./xmlchange PROJECT=UCUB0184,JOB_WALLCLOCK_TIME="12:00:00",JOB_QUEUE="regular"
./xmlquery PROJECT,JOB_QUEUE,JOB_WALLCLOCK_TIME
```

## Pass the project code when creating or building

You can also give the project code directly to the commands that need it:

```bash
./create_newcase --case ~/cases/case01 --compset B1850 --res f19_g17 --project UCUB0184
qcmd -A UCUB0184 -- ./case.build
```

More discussion of this issue is on the CESM forum:
https://bb.cgd.ucar.edu/cesm/threads/how-to-change-project-code.5846/
