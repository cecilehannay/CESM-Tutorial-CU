# Where to find output, log and timing files

| Question | Where to look |
|---|---|
| My run completed, where is my output? | **archive directory** |
| My run crashed, where can I find info? | **log files** in the run directory |
| I need to estimate resources for my run | **timing files** in the case directory |


## Overview of CESM directories: timing files

Timing files can be found in `CASE/timing` once a run completes.

<div class="alert alert-info">
When you need to estimate resources, do a short (e.g., 5-day) run and look at the timing
files to see the model throughput.
</div>
