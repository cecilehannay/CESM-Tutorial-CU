# Log files and history files

## Log files

- While the run is in progress, log files live in the **run directory** (`$RUNDIR`).
- Once the run completes, they are moved to `CASE/logs`.

<div class="alert alert-info">
If the run crashes, examine the log files in the run directory <code>$RUNDIR</code> — this
is where to find information about why your run crashed. See also the
<a href="../09.Troubleshooting/9.1_log_files_and_crashes.ipynb">Troubleshooting</a> chapter.
</div>


![Overview of CESM directories with log files](../../images/cesm_directories_log_files.png)

## History files

- While the run is in progress, history (output) files live in the **run directory**
  (`$RUNDIR`).
- Once the run completes, they are moved to the **archive directory**.
