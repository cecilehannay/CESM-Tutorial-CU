# Basic Unix commands

CESM is controlled primarily from a Unix terminal. You do not need to be a Unix expert, but you should be comfortable navigating directories and working with files.

:::{note}
Unix commands and filenames are case-sensitive. For example, `Case01` and `case01` are different names.
:::

## Understanding paths

A **path** describes the location of a file or directory.

| Symbol | Meaning |
|---|---|
| `~` | Your home directory |
| `.` | Your current directory |
| `..` | The directory one level above your current directory |
| `/` | The top of the directory system or the separator between directory names |

For example:

```bash
~/cases
```

means the `cases` directory inside your home directory.

An **absolute path** begins with `/` and gives the complete location:

```bash
/glade/derecho/scratch/$USER
```

A **relative path** describes a location starting from your current directory:

```bash
../case02
```

## Find your current location with `pwd`

The `pwd` command displays the name of the current working directory, which is the directory in which you are currently located. When you are unsure where you are, use `pwd`.

```bash
pwd
```

This displays the location of the current working directory, such as `/glade/u/home/$USER`.

## Change directories with `cd`

The `cd` command moves you from one directory to another.

```bash
cd /glade/u/home/$USER/cases
```

changes the current directory to `/glade/u/home/$USER/cases`.

```bash
cd ~
```

brings you back to your home directory.

## List files with `ls`

The `ls` command lists the files and directories in the current directory or in a directory you name.

```bash
ls
ls ~/cases
```

Useful options:

- `ls -l` shows more information: permissions, owner, file size, and the time of the last change.
- `ls -a` also shows hidden files, whose names begin with a dot (such as `~/.bashrc`).
- `ls -ltr` lists files sorted by modification time, newest last. This is useful when your model run crashes and you want to see which log file was updated most recently.

## Copy files or directories with `cp`

The `cp` command copies files or directories from a source location to a destination location.

```bash
cp file1 file2
```

copies the contents of `file1` to `file2`.

```bash
cp -r directory1 directory2
```

copies the contents of `directory1` to `directory2`. The `-r` (recursive) option is needed to copy a directory and everything inside it.

## Remove files or directories with `rm`

The `rm` command deletes files or directories. Be cautious, as the operation is irreversible.

```bash
rm file1
```

deletes the file `file1`.

```bash
rm -rf directory1
```

removes the directory `directory1` and all of its contents.

## Move or rename files with `mv`

The `mv` command moves or renames files and directories.

```bash
mv file1 file2
```

renames `file1` to `file2`.

```bash
mv file1 directory1
```

moves `file1` into the directory `directory1`.

## View a text file with `less`

The `less` command lets you read a text file one screen at a time without modifying it.

```bash
less file1
```

Useful keys inside `less` include:

| Key | Action |
|---|---|
| `Space` | Move forward one screen |
| `b` | Move backward one screen |
| `/word` | Search for `word` |
| `n` | Go to the next search result |
| `q` | Quit |

The older `more` command performs a similar function, but `less` provides more options.

```bash
more file1
```

To display only the beginning of a file, use:

```bash
head file1
```

To display the end of a file, use:

```bash
tail file1
```

This is particularly useful for examining CESM log files.

## Search inside files with `grep`

The `grep` command searches for lines containing a specified pattern within one or more text files.

```bash
grep hello file.txt
```

searches for the string `hello` in `file.txt` and displays all lines containing that string.

To search for a string in all the files inside a directory, use the `-r` (recursive) option:

```bash
grep -r hello directory1
```

## Select multiple files with wildcards (`*`)

The symbol `*` is a wildcard: it stands for any combination of characters. For example, to search for the string `hello` only in files that end in `.txt`:

```bash
grep hello *.txt
```

This searches for `hello` in all files that end with the `.txt` extension.

## Use environment variables

An environment variable stores a value that can be reused in commands. Its name is preceded by `$`.

Display your Derecho username:

```bash
echo $USER
```

Display your home-directory path:

```bash
echo $HOME
```

Display your CESM project:

```bash
echo $PROJECT
```

If `CASE` has been set to `case01`, this command:

```bash
cd ~/cases/$CASE
```

is equivalent to:

```bash
cd ~/cases/case01
```
