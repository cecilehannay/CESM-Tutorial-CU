# 9.2 Project code issues

If Derecho's batch system doesn't recognize your project number (for example,
`case.build` or `case.submit` complain about a missing/invalid project), you can check and
set the project explicitly with `xmlquery`/`xmlchange`:

```bash
./xmlquery PROJECT,JOB_QUEUE,JOB_WALLCLOCK_TIME
./xmlchange PROJECT=UCUB0184,JOB_WALLCLOCK_TIME="12:00:00",JOB_QUEUE="regular"
./xmlquery PROJECT,JOB_QUEUE,JOB_WALLCLOCK_TIME
```

When building, pass the project code to `qcmd` with `-A`:
```bash
qcmd -A UCUB0184 -- ./case.build
```

More discussion of this issue on the CESM forum:
https://bb.cgd.ucar.edu/cesm/threads/how-to-change-project-code.5846/
