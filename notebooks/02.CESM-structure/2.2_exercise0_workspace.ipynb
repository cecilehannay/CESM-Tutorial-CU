# Exercise 0a: Prepare your directories

For this tutorial, you need two directories in your home directory:

* `~/code` for the CESM source code
* `~/cases` for your CESM cases

![Overview of CESM directories, Preparing your space](../../images/02.prepare_your_space.png)

:::{admonition} Exercise 0a: Prepare your directories
:class: exercise
1. Create the directories `~/code` and `~/cases`.
2. Check that both directories exist.
3. Practice moving between the directories that already exist on Derecho.
:::

Create them with:

```bash
mkdir -p ~/code
mkdir -p ~/cases
```

The `-p` option prevents an error if a directory already exists.

Check that both directories are present:

```bash
ls -ld ~/code ~/cases
```

You do **not** need to create the other three directories:

* The input-data directory already exists on Derecho.
* CESM will create the build/run directory.
* CESM will create the archive directory.

You have now prepared the two directories that you will manage yourself. In the next sections, you will download CESM into `~/code` and create your first experiment in `~/cases`.
