# Exercise: Prepare your workspace

For this tutorial, you need two directories in your home directory:

* `~/code` for the CESM source code
* `~/cases` for your CESM cases

Create them with:

```bash
mkdir -p ~/code
mkdir -p ~/cases
```

The `-p` option prevents an error if a directory already exists.

Check that both directories are present:

```bash
ls -ld ~/code ~/cases
```

You do **not** need to create the other three directories:

* The input-data directory already exists on Derecho.
* CESM will create the build/run directory.
* CESM will create the archive directory.

You have now prepared the two directories that you will manage yourself. In the next sections, you will download CESM into `~/code` and create your first experiment in `~/cases`.


## Now pratice to navigate between directory

Assuming a case named `case01`, here is the command to reach each of the five directories:

| Directory | Command to get there |
| --- | --- |
| CESM code | `cd ~/code/release-cesm2.1.5` |
| Input data | `cd /glade/campaign/cesm/cesmdata/cseg/inputdata` |
| Case | `cd ~/cases/case01` |
| Build/run | `cd /glade/derecho/scratch/$USER/case01/run` |
| Archive | `cd /glade/derecho/scratch/$USER/archive/case01` |

Try the ones that already exist for you:

- `cd ~/code` (you just created it above)
- `cd ~/cases` (you just created it above)
- `cd /glade/campaign/cesm/cesmdata/cseg/inputdata` (this one already exists on Derecho for everyone)

The **case**, **build/run**, and **archive** directories for `case01` don't exist yet — you'll
create them in the next chapter when you download the CESM code and create your first case.
Come back to this table as a reference once you do.
