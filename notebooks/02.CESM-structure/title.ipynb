# Navigating the CESM maze

A CESM experiment on derecho does not live in a single directory. Its files are spread across five different directories, each with a specific purpose.

At first, this can feel like a maze, especially because the same case name appears in several places. In this section, you will learn what belongs in each directory, how the directories are connected, and when you will use them. Knowing your way around these five directories is an important step toward becoming a CESM expert.
