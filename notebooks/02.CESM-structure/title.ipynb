# Navigating the CESM maze

A CESM experiment on dercho does not live in a single directory. Its files are spread across five different directories, each with a specific purpose.

In this section, we introduce the five directories you will need to navigate between when
you run CESM ("the CESM maze"), and how to become a CESM expert by knowing your way around
them.

At first, this can feel like a maze—especially because the same case name appears in several places. The goal of this section is to help you understand what each directory contains and when you will use it.
