# Understanding the five directories

When you run CESM, you move between five directories:

![The five CESM directories](../../images/cesm_directories_five.png)

## What do `$USER` and `$CASE` mean?

- `$USER` is your Derecho username.
- `$CASE` is the name you give your experiment (your "case").
- `~` is a shortcut for your home directory, `/glade/u/home/$USER`.

Wherever you see `$USER` or `$CASE`, substitute your own username and case name. For
example, if your case is named `case01`, the case directory `~/cases/$CASE` is `~/cases/case01`.

```{important}
If your case is named `case01`, that name shows up in the **case**, **build/run**, and
**archive** directories — but these are three different directories, not copies of one
another. The case directory holds the instructions for the experiment, the build/run
directory is where CESM actually executes it, and the archive directory holds the organized
results. Always check the full path so you know which one you're in.
```

Don't worry if these paths still look complicated — we'll visit these directories repeatedly
throughout the tutorial.
