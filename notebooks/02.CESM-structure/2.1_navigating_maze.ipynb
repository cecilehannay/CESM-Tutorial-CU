# Understanding the five directories

The five directories are:

1. CESM code
2. Input data
3. Case
4. Build/run
5. Archive

You do not need to memorize every path immediately. For now, concentrate on understanding the purpose of each directory.

| Directory      | What is it used for?                                                                                                                                                 | Typical location                               |
| -------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------- | ---------------------------------------------- |
| **CESM code**  | Contains the CESM source code and the CIME tools used to create cases. You download the code once and can use it for many experiments.                               | `~/code/release-cesm2.1.5`                     |
| **Input data** | Contains large files needed to start and run CESM, including grids, forcing data, and initial conditions. This is a shared directory that already exists on Derecho. | `/glade/campaign/cesm/cesmdata/cseg/inputdata` |
| **Case**       | Contains the settings and commands for one experiment. This is where you configure, build, submit, and manage your case.                                             | `~/cases/$CASE`                                |
| **Build/run**  | Contains the CESM executable and the files used while the model is running. CESM creates this directory on scratch space.                                            | `/glade/derecho/scratch/$USER/$CASE/run`       |
| **Archive**    | Contains output and restart files that CESM organizes after a successful run.                                                                                        | `/glade/derecho/scratch/$USER/archive/$CASE`   |

## The same case name appears in several places

This is one of the most confusing parts for new CESM users.

Suppose your case is named `case01`. You will see that name in several directories:

| Location          | Example                                       |
| ----------------- | --------------------------------------------- |
| Case directory    | `~/cases/case01`                              |
| Run directory     | `/glade/derecho/scratch/$USER/case01/run`     |
| Archive directory | `/glade/derecho/scratch/$USER/archive/case01` |

These directories belong to the **same experiment**, but they are not copies of one another:

* The **case directory** contains the instructions for the experiment.
* The **run directory** is where CESM performs the simulation.
* The **archive directory** stores the organized results.

```{important}
The same case name does not mean the same directory. Always check the full path so that you know where you are.
```

## What do `$USER`, `$CASE`, and `~` mean?

The paths above use a few shortcuts:

* `$USER` represents your Derecho username.
* `$CASE` represents the name of your CESM experiment.
* `~` represents your home directory: `/glade/u/home/$USER`.

For example, if your username is `student1` and your case is `case01`:

```bash
/glade/derecho/scratch/$USER/$CASE/run
```

represents:

```bash
/glade/derecho/scratch/student1/case01/run
```

## Moving between the directories

Once `$CASE` has been set to your case name, you can use these commands:

| Directory  | Command                                           |
| ---------- | ------------------------------------------------- |
| CESM code  | `cd ~/code/release-cesm2.1.5`                     |
| Input data | `cd /glade/campaign/cesm/cesmdata/cseg/inputdata` |
| Case       | `cd ~/cases/$CASE`                                |
| Build/run  | `cd /glade/derecho/scratch/$USER/$CASE/run`       |
| Archive    | `cd /glade/derecho/scratch/$USER/archive/$CASE`   |

Do not worry if these paths still look complicated. We will visit the directories repeatedly during the tutorial.
