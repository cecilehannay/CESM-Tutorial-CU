# Understanding the five directories

When you run CESM, you move between five directories as representated in the figure and the table below.  

In the Figure and the Table:
- `$USER` is your Derecho username.
- `$CASE` is the name you give your experiment (your "case").
- `~` is a shortcut for your home directory, `/glade/u/home/$USER`.

Wherever you see `$USER` or `$CASE`, substitute your own username and case name. For example, if your case is named `case01`, the case directory `~/cases/$CASE` is `~/cases/case01`.

![The five CESM directories](../../images/02.cesm_directories_five.png)

The table below provides a description of the five directories. You do not need to memorize all the paths. The important thing is to understand the purpose of each directory.

| Directory | What is it, and how will you use it? | Typical location on Derecho |
|---|---|---|
| **1. CESM code** | Contains the CESM source code and the CIME scripts used to create new experiments. You download the code once and can use it to create many cases. | `~/code/release-cesm2.1.5` |
| **2. Input data** | Contains the grids, forcing data, and initial-condition files needed to run CESM. This is a large, shared directory that already exists on Derecho. CESM reads files from it, but you normally do not modify anything there. | `/glade/campaign/cesm/cesmdata/cseg/inputdata` |
| **3. Case** | The control center for one CESM experiment. It contains the experiment settings, user modifications, and commands for setting up, building, and submitting the simulation. This is the directory where you will do most of your work. | `~/cases/$CASE` |
| **4. Build/run** | The working area used to build and run the model. The `bld` directory contains the executable, while the `run` directory contains the files CESM needs during the simulation, along with output, restart, and log files. CESM creates these directories automatically on scratch space. | Build: `/glade/derecho/scratch/$USER/$CASE/bld`<br>Run: `/glade/derecho/scratch/$USER/$CASE/run` |
| **5. Archive** | Contains output, restart, and log files organized by CESM after a run. This is also called the **short-term archive**. CESM creates this directory automatically when archiving is enabled. | `/glade/derecho/scratch/$USER/archive/$CASE` |



Don't worry if these paths still look complicated — we'll visit these directories repeatedly
throughout the tutorial.

> **The most important point:** The same case name, represented by `$CASE`, appears in the case, build/run, and archive paths. These directories all belong to the same experiment, but they contain different files and serve different purposes.
>
> For example, if `$CASE` is `case01`, then `~/cases/case01`, `/glade/derecho/scratch/$USER/case01/run`, and `/glade/derecho/scratch/$USER/archive/case01` are three different directories for the same CESM experiment.



```{important}
If your case is named `case01`, that name shows up in the **case**, **build/run**, and
**archive** directories — but these are three different directories, not copies of one
another. The case directory holds the instructions for the experiment, the build/run
directory is where CESM actually executes it, and the archive directory holds the organized
results. Always check the full path so you know which one you're in.
```



The table below provides a map of the five directories used in this tutorial. You do not need to memorize all the paths now. The important thing is to understand the purpose of each directory.

| Directory | Typical location on Derecho and purpose |
|---|---|
| *@CESM code** | `~/code/release-cesm2.1.5`<br><br>Contains the CESM source code and the CIME scripts used to create new experiments. You download the code once and can use it to create many cases. |
| **Input data** | `/glade/campaign/cesm/cesmdata/cseg/inputdata`<br><br>Contains the grids, forcing data, and initial-condition files needed to run CESM. This shared directory already exists on Derecho. CESM reads files from it, but you normally do not modify anything there. |
| **Case** | `~/cases/$CASE`<br><br>The control center for one CESM experiment. It contains the experiment settings, user modifications, and commands for setting up, building, and submitting the simulation. This is where you will do most of your work. |
| **Build/run** | **Build:** `/glade/derecho/scratch/$USER/$CASE/bld`<br>**Run:** `/glade/derecho/scratch/$USER/$CASE/run`<br><br>The working area used to build and run CESM. The `bld` directory contains the executable. The `run` directory contains the files needed during the simulation, along with output, restart, and log files. CESM creates both directories automatically on scratch space. |
| **Archive** | `/glade/derecho/scratch/$USER/archive/$CASE`<br><br>Contains output, restart, and log files organized by CESM after a run. This is also called the **short-term archive**. CESM creates this directory automatically when archiving is enabled. |

> **The most important point:** The same case name, represented by `$CASE`, appears in the case, build/run, and archive paths. These directories all belong to the same experiment, but they contain different files and serve different purposes.
