# Understanding the five directories

When you run CESM, you move between five directories. You do not need to memorize every
path immediately — for now, focus on what each directory is for. The examples below assume
your case is named `case01`; wherever you see `$USER` and `$CASE`, substitute your own
username and case name.

| Directory | What is it used for? | Example path (case `case01`) | Command to get there |
| --- | --- | --- | --- |
| **CESM code** | Contains the CESM source code and the CIME tools used to create cases. You download it once and reuse it for many experiments. | `~/code/release-cesm2.1.5` | `cd ~/code/release-cesm2.1.5` |
| **Input data** | Large shared files needed to run CESM: grids, forcing data, initial conditions. Already exists on Derecho — you never create this yourself. | `/glade/campaign/cesm/cesmdata/cseg/inputdata` | `cd /glade/campaign/cesm/cesmdata/cseg/inputdata` |
| **Case** | The settings and scripts for one experiment. Where you configure, build, submit, and manage your case. | `~/cases/case01` | `cd ~/cases/$CASE` |
| **Build/run** | The CESM executable and the files used while the model is running. CESM creates this on scratch space. | `/glade/derecho/scratch/$USER/case01/run` | `cd /glade/derecho/scratch/$USER/$CASE/run` |
| **Archive** | Output and restart files, organized by CESM after a successful run. | `/glade/derecho/scratch/$USER/archive/case01` | `cd /glade/derecho/scratch/$USER/archive/$CASE` |

![The five CESM directories](../../images/cesm_directories_five.png)

```{important}
Notice that the case name `case01` shows up in the **case**, **build/run**, and **archive**
paths above — but these are three different directories, not copies of one another. The case
directory holds the instructions for the experiment, the build/run directory is where CESM
actually executes it, and the archive directory holds the organized results. Always check the
full path so you know which one you're in.
```

`~` is a shortcut for your home directory (`/glade/u/home/$USER`); `$USER` and `$CASE` are
shortcuts for your username and case name. Don't worry if these paths still look complicated
— we'll visit these directories repeatedly throughout the tutorial.
