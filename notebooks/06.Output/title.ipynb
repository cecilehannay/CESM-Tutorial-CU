# Output, log, and timing files

Once a run finishes (or crashes), you need to know where to look. This chapter answers three
questions you will ask again and again:

- My run completed. Where is my output?
- My run crashed. Where do I find out why?
- How long will a longer run take, and how many submissions do I need?

You will learn which of the five CESM directories holds the answer to each question, then
practice reading timing files and looking at real model output.
