# Output, log, and timing files

Once a run finishes (or crashes), you need to know where to look. CESM writes three kinds
of files you will care about:

- **Timing files**: how fast the model ran. You use them to plan longer runs.
- **Log files**: what each component printed while it ran. This is where you find out why a
  run crashed.
- **History files**: the model output (temperature, precipitation, winds, ...), stored as
  netCDF files.

Each kind ends up in a different one of the five CESM directories:

| Question | Where to look |
|---|---|
| How long will a longer run take, and how many submissions do I need? | **timing files** in the case directory |
| My run crashed. Where can I find out why? | **log files** in the run directory |
| My run completed. Where is my output? | **history files** in the archive directory |

In this chapter, you will first read a timing file and plan a long run (Exercise 3), then
find your log and history files and open real model output (Exercise 4).

In the examples, `b1850_hw` is the case from the homework. If you use a different case name,
replace it.
