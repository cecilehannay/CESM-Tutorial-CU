# Exercise 3: Estimate the run length and number of submissions

In this exercise, you use the timing file from your homework run to plan a much longer
simulation. If you did not finish the homework, you can use `case01` from Exercise 2 instead.

:::{admonition} Exercise 3: Plan a 30-year run
:class: exercise
Look up the model throughput of `b1850_hw`. Then decide which values of `STOP_OPTION`,
`STOP_N`, and `RESUBMIT` you would use to run **30 years**, assuming each submission is
allowed at most **12 hours** of wall-clock time.

Two definitions you need:

- **Model throughput**: the number of simulated model years completed per day of wall-clock
  time.
- **Maximum wall-clock time**: the longest a single submission is allowed to run on Derecho
  before the scheduler stops it.
:::

## Steps

1. Go to the timing directory of your case:

   ```bash
   cd ~/cases/b1850_hw/timing
   ```

2. Open the timing file (`less` works well) and find the line that looks like:

   ```text
   Overall Metrics:
       Model Throughput: XXX simulated years/day
   ```

3. From the throughput and the 12-hour limit, work out how many years fit in one submission.
4. Choose `STOP_OPTION`, `STOP_N`, and `RESUBMIT` so that the total adds up to 30 years. Remember
   that `RESUBMIT` counts the resubmissions *after* the first submission.

:::{note}
If your homework run did not complete, you can look at example timing data in
`/glade/u/home/tliefer/cases/case01`.
:::

:::{admonition} Solution
:class: solution dropdown
The timing file reports something like:

```text
Overall Metrics:
    Model Throughput: 14 simulated years/day
```

To be safe, plan on about 10 model years per day (queues and file systems are never perfectly
fast). In 12 hours, that is about 5 years per submission.

To run 30 years, you need 6 submissions of 5 years each: the first submission plus 5
resubmissions.

```text
STOP_N      = 5
STOP_OPTION = nyears
RESUBMIT    = 5
```
:::

## How to check it worked

There is nothing to run in this exercise. Check your answer against the solution above, and
make sure you can explain why `RESUBMIT` is 5 and not 6.
