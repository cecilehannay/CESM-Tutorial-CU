# Timing files


When a run completes, CESM writes **timing files** to the `timing/` subdirectory of the
case directory:

```bash
ls ~/cases/b1850_hw/timing
```

![Overview of CESM directories, timing files](../../images/06.cesm_directories_timing_files.png)


## Two different clocks

A timing file is confusing until you notice that it talks about two different clocks:

| | Model time | Wall-clock time |
|---|---|---|
| What it measures | time inside the climate model | real time passing on Derecho |
| The question it answers | "How much climate did we simulate?" | "How long did the computer work?" |
| Example | 1 month | 5 minutes |

Say a one-month run takes 5 minutes on Derecho. The model simulated one month of climate
(model time) while the computer worked for 5 minutes (wall-clock time). Fortunately, CESM is
faster than the real climate.

The **model throughput** connects the two clocks. It is the number of simulated model years
completed per wall-clock day, for example 25 simulated years per day.

The other number you need is the **maximum wall-clock time**: on Derecho no single job may run
longer than 12 hours. Put the two together and the timing number becomes useful:

```text
25 simulated years  =  24 wall-clock hours
12.5 simulated years  =  12 wall-clock hours      <- theoretical maximum per submission
```

But never plan a 12-hour job to finish at 11:59:59. Keep a safety margin and plan on about
10 model years per submission.


## What is inside a timing file

Open the timing file with `less` and look for the **model throughput**.

```text
Overall Metrics:
    Model Throughput: XXX simulated years/day
```

This is the number you need to plan a long experiment. Together with the maximum wall-clock
time of a single submission on Derecho, it tells you how many model years fit in one
submission, and therefore which values of `STOP_N` and `RESUBMIT` to use.

:::{tip}
Before launching a long experiment, do a short run first (a few days or one month), read the
throughput from its timing file, and plan the real run from that. The next exercise does
exactly this.
:::
