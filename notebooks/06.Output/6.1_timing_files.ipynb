# Timing files


When a run completes, CESM writes **timing files** to the `timing/` subdirectory of the
case directory:

```bash
ls ~/cases/b1850_hw/timing
```

![Overview of CESM directories, timing files](../../images/06.cesm_directories_timing_files.png)


## What is inside a timing file

Open the timing file with `less` and look for the **model throughput**: how many simulated
years the model completes in one day of wall-clock time.

```text
Overall Metrics:
    Model Throughput: XXX simulated years/day
```

This is the number you need to plan a long experiment. Together with the maximum wall-clock
time of a single submission on Derecho, it tells you how many model years fit in one
submission, and therefore which values of `STOP_N` and `RESUBMIT` to use.

:::{tip}
Before launching a long experiment, do a short run first (a few days or one month), read the
throughput from its timing file, and plan the real run from that. The next exercise does
exactly this.
:::
