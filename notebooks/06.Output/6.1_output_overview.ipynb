# Where to find output, log, and timing files

CESM writes three kinds of files you will care about: **history files** (the model output),
**log files** (what the model printed while it ran), and **timing files** (how fast it ran).
Each kind ends up in a different one of the five directories.

| Question | Where to look |
|---|---|
| My run completed. Where is my output? | **history files** in the archive directory |
| My run crashed. Where can I find out why? | **log files** in the run directory |
| How should I set up a longer run? | **timing files** in the case directory |

In the examples below, `b1850_hw` is the case from the homework. If you use a different case
name, replace it.

## History files: the model output

A **history file** is a netCDF file containing model output such as temperature, precipitation,
or winds, averaged over a period of time. While the run is in progress, history files are
written to the run directory. When the run completes, CESM moves them to the **archive
directory**, sorted by component:

```text
/glade/derecho/scratch/$USER/archive/b1850_hw/atm/hist/b1850_hw.cam.h0.0001-01.nc
```

The file name tells you what is inside:

| Part | Meaning |
|---|---|
| `b1850_hw` | the case name |
| `cam` | the component that wrote it (CAM is the atmosphere model) |
| `h0` | the history "stream"; `h0` is the default monthly-average output |
| `0001-01` | the model year and month (year 1, January) |

Other components use the same pattern: `clm2` for land, `pop` for ocean, `cice` for sea ice.

## Log files: what happened during the run

Each component writes a **log file** (`cesm.log.*`, `atm.log.*`, `ocn.log.*`, ...). While the
model is running, the logs are in the run directory. After a successful run, CESM moves them to
the `logs/` subdirectory of the case directory. If a run crashes, the logs usually stay in the
run directory, which is where you should look first. The
[Troubleshooting](../09.Troubleshooting/9.1_log_files_and_crashes.ipynb) chapter shows how to
read them.

## Timing files: how fast the model runs

When a run completes, CESM writes **timing files** to the `timing/` subdirectory of the case
directory:

```bash
ls ~/cases/b1850_hw/timing
```

The most useful number inside is the **model throughput**: how many simulated years the model
completes in one day of wall-clock time. You use it to decide how long each submission can be
and how many submissions a long experiment needs.

:::{tip}
Before launching a long experiment, do a short run first (a few days or one month), read the
throughput from its timing file, and plan the real run from that. The next exercise does
exactly this.
:::
