# Exercise 4: Look at model output

In this exercise, you open a real CESM history file and look at what is inside. Use the
monthly output from your homework case `b1850_hw`. If you did not finish the homework, use
`case01` from Exercise 2, which also produced one monthly file.

:::{admonition} Exercise 4: Open a monthly history file
:class: exercise
1. Find the January history file from the atmosphere model in your archive directory.
2. List the variables it contains with `ncdump -h`.
3. Open it in `ncview` and look at a few fields.
:::

## Steps

1. Go to the atmosphere history directory in the archive:

   ```bash
   cd /glade/derecho/scratch/$USER/archive/b1850_hw/atm/hist
   ls
   ```

   You should see a file named `b1850_hw.cam.h0.0001-01.nc`.

2. Print the file's header. `ncdump -h` lists every variable and its metadata (units, long
   name, dimensions) without printing the data itself:

   ```bash
   ncdump -h b1850_hw.cam.h0.0001-01.nc
   ```

3. Open the file in `ncview`, an interactive viewer that draws maps of each variable. It
   needs a graphical connection, so run it from a terminal where you logged in with
   `ssh -X`, not from JupyterHub:

   ```bash
   module load ncview
   ncview b1850_hw.cam.h0.0001-01.nc
   ```

:::{note}
If your run did not complete, you can look at example output in
`/glade/derecho/scratch/amiller/archive/case01/atm/hist`.
:::

## What you should see

`ncdump -h` prints a long list of variables. For example, `TREFHT` is the air temperature
near the surface and `PRECT` is the total precipitation rate.

![ncdump -h example output](../../images/ncdump_example.png)

`ncview` opens a small control window; click a variable name (for example `TREFHT`) to draw
its map.

![ncview example](../../images/ncview_example.png)
