## Optional: look at the same file with Python

:::{warning}

For real analysis you will want Python rather than `ncview`. The
[`xarray`](https://docs.xarray.dev/) library opens a CESM history file in one line and can
plot any variable in one more.

::::{admonition} Show the Python version
:class: note dropdown

Run this in a Jupyter notebook on Derecho or Casper (for example through
[NCAR OnDemand](https://ondemand.hpc.ucar.edu/pun/sys/dashboard), with a Python kernel that
has `xarray` and `matplotlib`, such as the "NPL" conda environment).

**Open your output file.** Point the path at one of your own monthly history files:

```python
import xarray as xr

path = "/glade/derecho/scratch/$USER/archive/b1850_hw/atm/hist/b1850_hw.cam.h0.0001-01.nc"
ds = xr.open_dataset(path)
ds
```

The last line prints a summary of every variable, dimension, and attribute in the file. It is
the Python equivalent of `ncdump -h`.

**Make a quick map.** Every `xarray` variable has a built-in `.plot()`:

```python
import matplotlib.pyplot as plt

ds["TREFHT"].isel(time=0).plot()
plt.title("Surface air temperature (TREFHT)")
plt.show()
```

**Try it yourself.**

- Plot a different variable, for example `PRECT` for precipitation.
- Make a global-mean time series instead of a map:
  ```python
  ds["TREFHT"].mean(["lat", "lon"]).plot()
  ```
- If your case wrote several monthly files, open them all at once with
  `xr.open_mfdataset("/path/to/*.nc")`.
