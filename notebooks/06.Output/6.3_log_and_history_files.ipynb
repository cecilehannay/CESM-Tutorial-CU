# Log files and history files


This page shows where log files and history files live at each stage of a run, and how
history files are named. In the examples, `b1850_hw` is the case from the homework. If you
use a different case name, replace it.


## Log files

Each component writes a **log file** (`cesm.log.*`, `atm.log.*`, `ocn.log.*`, ...) with
what it printed while it ran.

- While the run is in progress, the log files are in the **run directory** (`$RUNDIR`).
- Once the run completes successfully, CESM moves them to `logs/` in the **case directory**.

If a run crashes, the logs stay in the run directory. That is the place to look for the error
message. The [Troubleshooting](../09.Troubleshooting/9.1_log_files_and_crashes.ipynb) chapter
explains how to search them.

## History files

A **history file** is a netCDF file containing model output such as temperature,
precipitation, or winds, averaged over a period of time.

- While the run is in progress, the history files are in the **run directory**.
- Once the run completes, CESM moves them to the **archive directory**, sorted by component
  (`atm/hist`, `lnd/hist`, `ocn/hist`, ...).

![Overview of CESM directories, history files](../../images/06.cesm_directories_history_files.png)

### How history files are named

```text
/glade/derecho/scratch/$USER/archive/b1850_hw/atm/hist/b1850_hw.cam.h0.0001-01.nc
```

The file name tells you what is inside:

| Part | Meaning |
|---|---|
| `b1850_hw` | the case name |
| `cam` | the component that wrote it (CAM is the atmosphere model) |
| `h0` | the history "stream"; `h0` is the default monthly-average output |
| `0001-01` | the model year and month (year 1, January) |

Other components use the same pattern: `clm2` for land, `pop` for ocean, `cice` for sea ice.
