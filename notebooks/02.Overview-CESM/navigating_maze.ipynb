# Becoming a CESM expert: navigating the CESM maze

![The CESM maze](../../images/cesm_maze.png)

When you run CESM, you have to navigate between **5 directories**:

1. CESM code
2. Inputdata directory
3. Case directory
4. Build/run directory
5. Archive directory

Each of these directories contains sub-directories. One of the goals of this tutorial is to
teach you how to navigate between these directories, so that "the maze" stops feeling like a
maze!


## The five directories, in practice

If your username is `$USER` and your case name is `$CASE`, you can navigate between the five
directories with:

| Directory | Command |
|---|---|
| CESM code | `cd ~/code/release-cesm2.1.5` |
| Inputdata directory | `cd /glade/campaign/cesm/cesmdata/cseg/inputdata` |
| Case directory | `cd ~/cases/$CASE` |
| Build/run directory | `cd /glade/derecho/scratch/$USER/$CASE/run` |
| Archive directory | `cd /glade/derecho/scratch/$USER/archive/$CASE` |

<div class="alert alert-info">
On Derecho, <code>~</code> is a shortcut for your home directory,
<code>/glade/u/home/$USER</code>. Timing files can be found in <code>CASE/timing</code> once a
run completes.
</div>

## Exercise 0: Prepare our space

Let's prepare our space by creating two directories in your home directory:
```
mkdir ~/code
mkdir ~/cases
```
The other directories we will need (build/run and archive directories) are created
automatically by CESM, or already exist for you.
