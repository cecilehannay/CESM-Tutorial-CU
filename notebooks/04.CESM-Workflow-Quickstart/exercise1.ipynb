# 4.3 Exercise 1

<div class="alert alert-info">

<strong>Exercise 1: Create, build and submit your first case</strong>

Use the 4 CESM commands to create a new case. Do the case setup, build and submit your job.
Use `qstat` to check that your job is running. Check that your run completes. Check which
files are in the archive directory.

</div>


```bash
# go into the scripts directory of the CESM code
cd ~/code/release-cesm2.1.5/cime/scripts

# (1) create a new case in the directory "cases" in your home directory
./create_newcase --case ~/cases/case01 --compset B1850 --res f19_g17

# go into the case you just created in the last step
cd ~/cases/case01/

# (2) setup your case
./case.setup

# (3) build the executable
qcmd -- ./case.build

# (4) submit your run
./case.submit
```


<div class="alert alert-warning">
<details>
<summary><font face="Times New Roman" color='blue'>If Derecho doesn't recognize your project number</font></summary><br>

If `create_newcase` or `case.build` complain about not knowing which project to charge, add
the project explicitly:
```bash
./create_newcase --case ~/cases/case01 --compset B1850 --res f19_g17 --project UCUB0184

cd ~/cases/case01/
./case.setup

qcmd -A UCUB0184 -- ./case.build

./case.submit
```
