# More info about compsets and grids

## Component sets (compsets)

A CESM **component set** (compset) defines which components are active, which are data
components, and which are stubs.

- **active**: the component is fully simulated
- **data**: the component reads prescribed data instead of being fully simulated
- **stub**: the component does nothing (a placeholder)

CESM is designed to be "plug and play": you can mix and match active, data and stub
components through the compset.

In this tutorial we use **B compsets** (fully-coupled model). Two examples:

- `B1850`: pre-industrial climate (representative of the year 1850). CO2 and other forcings
  cycle over a single year.
- `BHIST`: historical run (20th century). Forcings vary year to year.

<div class="alert alert-info">
We use <code>B1850</code> for the exercises rather than <code>BHIST</code> because its forcing
is fixed/cycling instead of a year-by-year timeseries — that makes it cheaper to spin up and
perfectly reproducible run to run, which is exactly what you want for a short tutorial
exercise. <code>BHIST</code> is what you'd reach for once you're doing a real historical or
present-day simulation.
</div>


## Grids (resolution)

The `--res` argument specifies the model resolution (grid). For example:
```bash
./create_newcase --case ~/cases/case01 --compset B1850 --res f19_g17
```
`f19_g17` means an approximately 2-degree finite-volume atmosphere/land grid (`f19`) combined
with the `gx1v7` ocean/ice grid (`g17`).

![Example model grid](../../images/grid_resolution.png)

Every grid also has a `long_name`, provided for future reference in case you become a CESM
power user — it is unlikely you will need it in this class.
