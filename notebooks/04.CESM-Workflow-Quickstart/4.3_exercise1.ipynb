# Exercise 1: Create, build, and submit your first case

In this exercise, you will apply the four-step CESM workflow introduced on the previous page.

## Your task

Create a CESM case with the following settings:

| Setting | Value |
|---|---|
| Case name | `case01` |
| Component set | `B1850` |
| Resolution | `f19_g17` |
| Case directory | `~/cases/case01` |

Then:

1. Set up the case.
2. Build the CESM executable.
3. Submit the simulation.
4. Use `qstat` to follow the job.
5. Confirm that the simulation completed successfully.
6. Examine the files in the archive directory.

Try completing the exercise using the workflow from the previous page. Open the solution below if you need help or want to check your commands.

```{dropdown} Show the solution

### Create the case

```bash
cd ~/code/release-cesm2.1.5/cime/scripts

./create_newcase \
    --case ~/cases/case01 \
    --compset B1850 \
    --res f19_g17
```

### Set up, build, and submit the case

```bash
cd ~/cases/case01

./case.setup

qcmd -- ./case.build

./case.submit
```

### Follow the job

```bash
qstat -u $USER
```

Run this command periodically. The main job will usually move from `Q` (queued) to `R` (running) before disappearing from the queue.

### Confirm that the run succeeded

After the job disappears from `qstat`, check the case status:

```bash
grep "case.run" CaseStatus
```

A successful simulation will include:

```text
case.run success
```

Check whether the output files were archived:

```bash
grep "st_archive" CaseStatus
```

When archiving is complete, you should see:

```text
st_archive success
```

### Examine the archive directory

```bash
ls /glade/derecho/scratch/$USER/archive/case01
```

You should see subdirectories containing model output, restart files, and log files.

```

```{admonition} If Derecho does not recognize your project code
:class: tip, dropdown

Normally, Derecho obtains your project code from the environment variables you set in the prerequisites.

If `create_newcase` does not recognize your project, add the project explicitly:

```bash
./create_newcase \
    --case ~/cases/case01 \
    --compset B1850 \
    --res f19_g17 \
    --project UCUB0184
```

If `qcmd` does not recognize your project when you build the model, use:

```bash
qcmd -A UCUB0184 -- ./case.build
```

See [Project-code issues](https://cecilehannay.github.io/CESM-Tutorial-CU/notebooks/09.Troubleshooting/9.2_project_code_issues.html) for additional help.
