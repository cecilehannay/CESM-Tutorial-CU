# Download the CESM Code

In this exercise, you will download CESM2.1.5 from GitHub into your `~/code` directory on Derecho.

A **GitHub repository** is an organized collection of files and their revision history. The CESM repository is public, so you do not need a GitHub account to download it.

> **Important:** These instructions download the CESM source code. They do not download the much larger CESM input-data collection. The input data are already available in a shared directory on Derecho.

## Step 1: Go to your code directory

First, make sure that your `~/code` directory exists:

```bash
mkdir -p ~/code
```

Move into that directory:

```bash
cd ~/code
```

Confirm your location:

```bash
pwd
```

The result should look like:

```text
/glade/u/home/<your_username>/code
```

## Step 2: Clone CESM2.1.5

There are many versions of CESM. For this tutorial, we will use:

```text
release-cesm2.1.5
```

Download this version with:

```bash
git clone -b release-cesm2.1.5 https://github.com/ESCOMP/CESM.git release-cesm2.1.5
```

This command does three things:

- `git clone` downloads a copy of the repository.
- `-b release-cesm2.1.5` selects the CESM version used in this tutorial.
- The final `release-cesm2.1.5` gives the new directory its name.

When the command finishes, move into the new directory:

```bash
cd release-cesm2.1.5
```

Confirm your location:

```bash
pwd
```

The result should look like:

```text
/glade/u/home/<your_username>/code/release-cesm2.1.5
```

## Step 3: Download the CESM components

The main CESM repository does not contain all the component-model code. CESM uses a script to download the correct versions of CAM, CLM, CICE, POP, CIME, and the other components.

Run:

```bash
./manage_externals/checkout_externals
```

This step may take several minutes and will produce many lines of output. That is normal.

## Step 4: Verify the download

Check the status of all the external components:

```bash
./manage_externals/checkout_externals -S
```

A successful checkout lists the component directories without status characters in front of them. The output will include directories such as:

```text
./cime
./components/cam
./components/cice
./components/clm
./components/pop
```

If you see a status such as `e-` before a component, that component was not downloaded correctly. Try running the download command again:

```bash
./manage_externals/checkout_externals
```

If the problem continues, ask an instructor for help.

## About the “detached HEAD” message

Git may display a message saying that you are in a **detached HEAD** state. This sounds alarming, but it is expected.

It simply means that you selected a specific, fixed CESM release rather than an active development branch. You can safely ignore this message for the tutorial.

## If the CESM directory already exists

If Git reports:

```text
fatal: destination path 'release-cesm2.1.5' already exists and is not an empty directory
```

do not run the clone command repeatedly. Check what is already in your code directory:

```bash
ls ~/code
```

If `release-cesm2.1.5` is already present, enter it and check its status:

```bash
cd ~/code/release-cesm2.1.5
./manage_externals/checkout_externals -S
```

Ask an instructor for help if you are unsure whether the existing download is complete.

## Fallback: Copy the course version

Use this option only if GitHub is unavailable and an instructor tells you to use the course copy.

First, confirm that `release-cesm2.1.5` does not already exist in your `~/code` directory:

```bash
ls ~/code
```

Then copy the prepared version:

```bash
mkdir -p ~/code
cd ~/code
cp -r /glade/u/home/jenkay/code/release-cesm2.1.5 .
```

Enter the copied directory and check it:

```bash
cd release-cesm2.1.5
./manage_externals/checkout_externals -S
```

## What you should have now

You should now have a complete copy of CESM2.1.5 at:

```text
~/code/release-cesm2.1.5
```

This is your **CESM code directory**. You will use it throughout the tutorial to create and run experiments.

For more detail, see the [official CESM2.1 download instructions](https://escomp.github.io/CESM/versions/cesm2.1/html/downloading_cesm.html).

```bash
# create a directory with the UNIX command mkdir
mkdir ~/code

# change to that directory
cd ~/code

# Download CESM in that directory
git clone https://github.com/ESCOMP/CESM.git release-cesm2.1.5
cd release-cesm2.1.5
git checkout release-cesm2.1.5
./manage_externals/checkout_externals
```

There are many versions (tags) of CESM. For this tutorial we use `release-cesm2.1.5`.

More about downloading the CESM code:
[CESM Tutorial: code overview](https://ncar.github.io/CESM-Tutorial/notebooks/basics/code_overview.html)


## Don't panic: the "scary" `git checkout` message

When you run `git checkout release-cesm2.1.5`, you will see a message about a "detached HEAD"
state. This looks alarming, but it isn't — you are safe.
