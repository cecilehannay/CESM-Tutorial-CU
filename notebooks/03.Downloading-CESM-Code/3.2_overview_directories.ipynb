# Overview of CESM directories: where are we so far?

So far, we have downloaded the CESM code. Let's take stock of where things live:

- **CESM code**: this is where the CESM code lives.
  - The `cime` directory contains all the scripts you need to run CESM.
  - The `components` directory contains all the CESM model code.
- **Inputdata directory**: this is where the CESM input data lives (topography, initial
  conditions, greenhouse gases, ...).
- **Case directory** (`~/cases`): this is where you will put all your cases (= simulations).
