# CESM directories after download

So far, we have downloaded the CESM code. Before creating your first experiment, let’s review which of the five CESM directories are already available and which CESM will create later.

- **CESM code**: this is where the CESM code lives.
  - The `cime` directory contains all the scripts you need to run CESM.
  - The `components` directory contains all the CESM model code.
- **Inputdata directory**: this is where the CESM input data lives (topography, initial
  conditions, greenhouse gases, ...).
- **Case directory** (`~/cases`): this is where you will put all your cases (= simulations).

# Where are we in the CESM maze?

You have completed the first step: **downloading the CESM source code**. Before creating your first experiment, let’s review which of the five CESM directories are already available and which CESM will create later.

![Overview of the CESM directories at this stage](https://cecilehannay.github.io/CESM-Tutorial-CU/_images/cesm_directories_early.png)

| Directory | Where are we now? |
|---|---|
| **CESM code** | `~/code/release-cesm2.1.5`<br><br>You downloaded this directory from GitHub. The `cime` directory contains the scripts and tools used to create and manage CESM experiments. The `components` directory contains the source code for CESM's atmosphere, land, ocean, sea-ice, and other model components. |
| **Input data** | `/glade/campaign/cesm/cesmdata/cseg/inputdata`<br><br>This shared directory already exists on Derecho. It contains files CESM needs, including grids, topography, forcing data, and initial conditions. You did not download these files, and you will normally only read from this directory. |
| **Case** | `~/cases/$CASE`<br><br>You created the parent directory `~/cases`, but you have not created an individual case yet. Each CESM experiment will have its own case directory inside `~/cases`. |
| Build/run** | **Not created yet.**<br><br>CESM will create the build and run directories on Derecho's scratch storage when you set up, build, and run your first case. |
| **Archive** | **Not created yet.**<br><br>CESM will create the archive directory after your simulation runs and its files are archived. |

In the next section, you will create your first CESM case. This will add the **case directory** to the maze. The **build/run** and **archive** directories will appear during later steps.
