# Prerequisites

Before beginning the tutorial, please make sure that:

- You can log in to **Derecho**. All exercises are designed to run on Derecho and will not work on Casper.
- You have access to a valid **project code**. Derecho uses this code to determine which project allocation should be charged for your jobs.
- You are familiar with basic **Unix commands**, including navigating directories and creating, copying, moving, and editing files.
- You know how to edit files using a **text editor** such as Emacs or Vi, or through JupyterLab.
- You can open, edit, and run a **Jupyter notebook**.

Because our class time is limited, these topics will not be covered during the tutorial. This prerequisite section provides basic instructions for accessing Derecho, along with a brief introduction to essential Unix commands.
