# Getting set up on Derecho

All CESM exercises in this tutorial must be completed on **Derecho**, **not Casper**. Being able to access Casper does not necessarily mean that you can access Derecho. You need to make sure you can connect to Derecho before the first class.

You can connect to Derecho from a **terminal**, through **NCAR OnDemand** (a web portal), or through **NCAR JupyterHub**. Whichever you choose, what you need in the end is a terminal running on Derecho.

## Option 1: Log in from a terminal

From a terminal on your computer, connect to Derecho with:
```bash
ssh -X $USER@derecho.hpc.ucar.edu
```
where you replace `$USER` with your UCAR username. Enter your UCAR password when prompted and approve the Duo authentication request.

For additional information, see the [NCAR Derecho documentation](https://ncar-hpc-docs.readthedocs.io/en/latest/compute-systems/derecho/).

## Option 2: Connect through NCAR OnDemand

[NCAR OnDemand](https://ondemand.hpc.ucar.edu/pun/sys/dashboard) is a web portal that gives you a terminal or a JupyterLab session on Derecho from your browser, with nothing to install. Bookmark it.

1. Go to [ondemand.hpc.ucar.edu](https://ondemand.hpc.ucar.edu/pun/sys/dashboard) and sign in with your UCAR username, password, and Duo authentication.
2. **For a terminal only:** open the **Clusters** menu at the top of the dashboard and choose the **Derecho** shell access entry. A terminal on a Derecho login node opens in a new browser tab. You are ready to go.
3. **For JupyterLab (notebooks and a terminal):** under **Interactive Apps**, click **Jupyter**, then fill in the form:
   - choose the **Derecho** option in the cluster drop-down (not Casper),
   - set the **Project account** to `UCUB0205`,
   - set the **Wall time** to about 2 hours, so the session lasts through the whole class.

   Click **Launch**, wait for the job to change from *Queued* to *Running*, then click **Connect to Jupyter**. Inside JupyterLab, open a terminal from the Launcher (**File → New → Terminal**).

:::{tip}
The project account and wall time only matter for the Jupyter option, which runs as a batch job. The shell-access terminal runs directly on a login node and needs neither.
:::

## Option 3: Connect through NCAR JupyterHub

You can also access Derecho through [NCAR JupyterHub](https://jupyterhub.hpc.ucar.edu). This older route is being replaced by OnDemand, but it still works:

1. Select **Production**.
2. Log in with your UCAR username and Duo authentication.
3. Choose **Derecho** when prompted to select a cluster.
4. Start JupyterLab and open a terminal.

See the [NCAR JupyterHub documentation](https://ncar-hpc-docs.readthedocs.io/en/latest/compute-systems/jupyterhub/) for more information.

## Set your project number

Derecho jobs must be associated with a project allocation.

For this class (**Climate Modeling Laboratory - 2026-27**), we use the project:

```text
UCUB0205
```

To make this project code available automatically each time you start a Bash session, add the following lines to your `~/.bashrc` file:

```bash
export PROJECT=UCUB0205
export PBS_ACCOUNT=${PROJECT}
```

These variables serve different purposes:

- `PROJECT` tells CESM which project to use.
- `PBS_ACCOUNT` tells the PBS scheduler which project allocation to charge.

If `~/.bashrc` does not already exist, your text editor will create it when you save the file.

Load the new settings:

```bash
source ~/.bashrc
```

Check that they were set correctly:

```bash
echo $PROJECT
echo $PBS_ACCOUNT
```

Both commands should return:

```text
UCUB0205
```

:::{important}
If you use Derecho for purposes other than this class, don't add these lines blindly to your
`~/.bashrc` — talk to us first so we can figure out the best way to set your project
number for this tutorial without disrupting your other work.
:::
