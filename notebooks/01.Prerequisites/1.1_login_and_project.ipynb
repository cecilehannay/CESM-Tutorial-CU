# Getting set up on Derecho

All CESM exercises in this tutorial must be completed on **Derecho**, not Casper. Being able to access Casper does not necessarily mean that you can access Derecho.

You can connect to Derecho from a terminal or through NCAR JupyterHub.

## Option 1: Log in from a terminal

From a terminal on you computer, connect to Derecho with:
```
ssh -X $USER@derecho.hpc.ucar.edu
```
Replace $USER with your UCAR username.

Enter your UCAR password when prompted and approve the Duo authentication request.

To confirm that you are connected to Derecho, run:

```bash
hostname
```

The result should identify a Derecho login node.

For additional information, see the [NCAR Derecho documentation](https://ncar-hpc-docs.readthedocs.io/en/latest/compute-systems/derecho/).

## Option 2: Connect through NCAR JupyterHub

You can also access Derecho through [NCAR JupyterHub](https://jupyterhub.hpc.ucar.edu):

1. Select **Production**.
2. Log in with your UCAR username and Duo authentication.
3. Choose **Derecho** when prompted to select a cluster.
4. Start JupyterLab and open a terminal.

See the [NCAR JupyterHub documentation](https://ncar-hpc-docs.readthedocs.io/en/latest/compute-systems/jupyterhub/) for more information.

> **Important:** Use the Derecho login nodes to edit files, configure CESM, and submit jobs. Do not run CESM directly on a login node. CESM simulations must be submitted to the compute nodes through PBS.


## Set your project number

Derecho jobs must be associated with a project allocation. 

For this class (Climate Modeling Laboaratory - 2025-26), we use:

```text
UCUB0184
```

If you want derecho to automatically know your project number. 

In `~/.bashrc`, add the lines:
```
export PROJECT=UCUB0184
export PBS_ACCOUNT=${PROJECT}
```

These variables serve different purposes:

- `PROJECT` tells CESM which project to use.
- `PBS_ACCOUNT` tells the PBS scheduler which project allocation to charge.

If `~/.bashrc` does not already exist, your text editor will create it when you save the file.

Load the new settings:

```bash
source ~/.bashrc
```

Check that they were set correctly:

```bash
echo $PROJECT
echo $PBS_ACCOUNT
```

Both commands should return:

```text
UCUB0184
```
