# Getting set up on Derecho

## Log in to Derecho 

Running CESM requires you to log in to **Derecho** (not Casper). 
<div class="alert alert-warning">
Being able to log in to Casper does not necessarily mean that you can log in to Derecho.
</div>

From a terminal, connect to Derecho with:
```
ssh -X $USER@derecho.hpc.ucar.edu
```
Replace $USER with your UCAR username.

You can also connect to Derecho from a terminal opened through the NCAR JupyterHub.
Alternatively, go to NCAR JupyterHub, select Production, log in with your UCAR credentials and Duo, and choose Derecho as the cluster. NCAR JupyterHub documentation

## Set your project number

PROJECT tells CESM which project to use.
PBS_ACCOUNT tells the PBS scheduler which allocation to charge.

In `~/.bashrc`, add the lines:
```
export PROJECT=UCUB0184
export PBS_ACCOUNT=${PROJECT}
```

If you don't have a file called `~/.bashrc`, create one with the two lines above.

Then run:
```
source ~/.bashrc
```

<div class="alert alert-warning">
If you use Derecho for purposes other than this class, don't add these lines blindly to your
<code>~/.bashrc</code> — talk to us first so we can figure out the best way to set your project
number for this tutorial without disrupting your other work.
</div>
