# Basic Unix Commands

## cd (Change Directory):

**Description**: 

The `cd` command is used to navigate from a directory to another directory. 

**Example**:

`cd /home/user/documents` - Changes the current directory to */home/user/documents*.

`cd ~` - Brings you to your *home* directory. 

<hr style="border: 1px solid black;">

## ls (List):

**Description**: 

The `ls` command is used to list the files and directories in the current directory or a specified directory. It provides information about the contents of the directory, such as filenames, permissions, sizes, and timestamps.

**Example**:

`ls` - Lists the files and directories in the current directory.

`ls -all /home/user/documents`  - Lists the contents of the */home/user/documents* directory showing detailed information about each file and directory.

There are other flags that can be used with `ls` to list the files/directories in different ways e.g., 

`ls -l` provides more information such as the permissions settings, the file size, the time of creation etc.

`ls -ltr` lists the files in reverse order of the time they were created.  This can be useful when your model run crashes and you want to see which log file was most recently created.

<hr style="border: 1px solid black;">


## pwd (Print Working Directory):

**Description**: 

The `pwd` command displays the name of the current working directory, which is the directory in which you are currently located. It helps you determine your current location in the directory tree.

**Example**:

`pwd` - Displays the location of the current working directory, such as `/home/user`.

<hr style="border: 1px solid black;">

## cp (Copy):

**Description**: 

The `cp` command copies files or directories from a source location to a destination location. 

**Example**: 

`cp file1 file2` - Copies the contents of *file1* to *file2*

`cp -r directory1 directory2` - Copies the contents of *directory1* to *directory2*.  The `-r` here is needed to recursively copy all the contents of the directory1 to directory2

<hr style="border: 1px solid black;">

## rm (Remove):

**Description**: 

The `rm` command deletes files or directories. Be cautious, as the operation is irreversible.

**Example**: 

`rm file1` - Deletes a file named *file1* 

`rm -rf directory1` - Removes a directory named *directory1* and all of its contents.

<hr style="border: 1px solid black;">

## mv (Move):

**Description**: 

The `mv` command moves or renames files and directories. It can also be used to move files from one directory to another.

**Example**: 

`mv file1 file2` - Renames *file1* to *file2*

`mv file1 directory1` - Moves *file1* into the directory *directory1*

<hr style="border: 1px solid black;">

## more/less (File Content Viewer):


**Description**: 

The command `more` allows you to view the content of a text file one page at a time. 
Similar to `more`, the command `less` lets you view the content of text files one page at a time. 

**Example**: 

`more file1`  displays the contents of a text file named file1.

<hr style="border: 1px solid black;">

## grep (Global Regular Expression Print):

**Description**: 

The command `grep` searches for lines containing a specified pattern within one or more text files. Useful for text searching and pattern matching.

**Example**: 

`grep hello file.txt` searches for the string *hello* in the *file.txt* file and displays all lines containing that string.

If you want to search for a string recursively within all files that are in the directory `directory1` you can use the `-r` flag i.e.,

`grep -r directory1`


## wildcards (*):

**Description**:

The symbol * can be used as a wildcard in unix.  The * can be used to represent any combination of characters.  For example, suppose you want to search for the string `hello` but only in files that end in `.txt`, you can use the following

`grep hello *.txt`

This will search for the string *hello* in all files that end with the .txt extension.

## Essential Unix commands

CESM is controlled primarily from a Unix terminal. You do not need to be a Unix expert, but you should be comfortable navigating directories and working with files.

> **Remember:** Unix commands and filenames are case-sensitive. For example, `Case01` and `case01` are different names.

## Understanding paths

A **path** describes the location of a file or directory.

| Symbol | Meaning |
|---|---|
| `~` | Your home directory |
| `.` | Your current directory |
| `..` | The directory one level above your current directory |
| `/` | The top of the directory system or the separator between directory names |

For example:

```bash
~/cases
```

means the `cases` directory inside your home directory.

An **absolute path** begins with `/` and gives the complete location:

```bash
/glade/derecho/scratch/$USER
```

A **relative path** describes a location starting from your current directory:

```bash
../case02
```

## Find your current location with `pwd`

The `pwd` command means **print working directory**. It displays the full path of your current directory.

```bash
pwd
```

Example output:

```text
/glade/u/home/student1/cases
```

When you are unsure where you are, use `pwd`.

## List files with `ls`

The `ls` command lists files and directories.

List the contents of your current directory:

```bash
ls
```

Show more information about each file:

```bash
ls -l
```

Show hidden files, whose names begin with `.`:

```bash
ls -a
```

Show detailed information with human-readable file sizes:

```bash
ls -lh
```

Sort files by modification time, with the most recently modified files at the bottom:

```bash
ls -ltr
```

This is useful when a CESM run stops and you want to identify the most recently created or modified log file.

You can also list the contents of another directory without moving into it:

```bash
ls -l ~/cases
```

## Change directories with `cd`

The `cd` command means **change directory**. It moves you from one directory to another.

Move to your home directory:

```bash
cd ~
```

Move into the `cases` directory in your home directory:

```bash
cd ~/cases
```

Move up one directory level:

```bash
cd ..
```

Return to the directory you were previously using:

```bash
cd -
```

A useful pattern is:

```bash
cd <directory>
pwd
ls
```

This moves you into a directory, confirms your location, and shows its contents.

## Create directories with `mkdir`

The `mkdir` command means **make directory**.

Create a directory called `cases`:

```bash
mkdir ~/cases
```

Use `-p` to avoid an error if the directory already exists:

```bash
mkdir -p ~/cases
```

Create several directories at once:

```bash
mkdir -p ~/code ~/cases
```

## Copy files with `cp`

The `cp` command copies a file. The original file remains unchanged.

```bash
cp <source> <destination>
```

Copy `file1` and name the copy `file2`:

```bash
cp file1 file2
```

Copy a file into another directory:

```bash
cp file1 ~/cases
```

Copy an entire directory and its contents:

```bash
cp -r directory1 directory2
```

The `-r` option means **recursive**: it copies the directory and everything inside it.

Use `-i` if you want Unix to ask before overwriting an existing file:

```bash
cp -i file1 file2
```

## Move or rename files with `mv`

The `mv` command moves or renames files and directories.

Rename `file1` as `file2`:

```bash
mv file1 file2
```

Move `file1` into `directory1`:

```bash
mv file1 directory1
```

Use `-i` to be asked before overwriting an existing file:

```bash
mv -i file1 file2
```

## Remove files with `rm`

The `rm` command permanently deletes files.

```bash
rm file1
```

> **Warning:** Files deleted with `rm` do not go to the Trash and are usually difficult or impossible to recover. Check the filename and your current directory before running the command.

Use `-i` to be asked for confirmation:

```bash
rm -i file1
```

To remove an empty directory, use:

```bash
rmdir directory1
```

To remove a directory and everything inside it, use `rm -r`:

```bash
rm -r directory1
```

Be especially careful with recursive removal. Do not use `rm -rf` unless you fully understand the command and have checked the complete path.

## View a text file with `less`

The `less` command lets you read a text file one screen at a time without modifying it.

```bash
less file1
```

Useful keys inside `less` include:

| Key | Action |
|---|---|
| `Space` | Move forward one screen |
| `b` | Move backward one screen |
| `/word` | Search for `word` |
| `n` | Go to the next search result |
| `q` | Quit |

The older `more` command performs a similar function, but `less` provides more options.

```bash
more file1
```

To display only the beginning of a file, use:

```bash
head file1
```

To display the end of a file, use:

```bash
tail file1
```

This is particularly useful for examining CESM log files.

## Search inside files with `grep`

The `grep` command searches for text inside one or more files.

Find every line containing `ERROR` in `cesm.log`:

```bash
grep "ERROR" cesm.log
```

Show the line numbers of matching lines:

```bash
grep -n "ERROR" cesm.log
```

Ignore differences between uppercase and lowercase letters:

```bash
grep -i "error" cesm.log
```

Search recursively through files inside a directory:

```bash
grep -r "ERROR" directory1
```

A recursive search can take a long time in a large directory. Do not run `grep -r` on large shared directories such as the CESM input-data directory.

## Select multiple files with wildcards

The `*` symbol is a **wildcard**. It represents any number of characters.

List every file ending in `.log`:

```bash
ls *.log
```

Search for `ERROR` in every log file:

```bash
grep "ERROR" *.log
```

List files whose names begin with `cesm`:

```bash
ls cesm*
```

Before copying, moving, or deleting files with a wildcard, use `ls` with the same pattern to check which files will be selected.

For example, check the files first:

```bash
ls *.log
```

Then remove them only if you are certain:

```bash
rm -i *.log
```

## Use environment variables

An environment variable stores a value that can be reused in commands. Its name is preceded by `$`.

Display your Derecho username:

```bash
echo $USER
```

Display your home-directory path:

```bash
echo $HOME
```

Display your CESM project:

```bash
echo $PROJECT
```

If `CASE` has been set to `case01`, this command:

```bash
cd ~/cases/$CASE
```

is equivalent to:

```bash
cd ~/cases/case01
```

## Useful terminal shortcuts

- Press **Tab** to complete a filename or directory name.
- Press the **Up Arrow** to reuse a previous command.
- Press **Ctrl+C** to stop a command that is running.
- Use `clear` to clear the terminal screen.
- Add `--help` after many commands to see available options:

```bash
ls --help
```
