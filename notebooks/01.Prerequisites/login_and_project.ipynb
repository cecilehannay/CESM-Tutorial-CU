# Logging in and setting up your project number

## JupyterHub / account issues

Running CESM requires you to log in to **Derecho** (not Casper).

If you cannot connect through the JupyterHub, you can use a terminal instead. The command to
log in to Derecho from a terminal is:
```
ssh -X $USER@derecho.hpc.ucar.edu
```

If you have an account problem (e.g., Duo not working, account locked), contact the UCAR
Help Desk:
- Email: help@ucar.edu
- Phone: (303) 497-2000

Please provide them with:
- Your username
- Your project number: `UCUB0184`


## Getting Derecho to know your project number

In `~/.bashrc`, add the lines:
```
export PROJECT=UCUB0184
export PBS_ACCOUNT=${PROJECT}
```

If you don't have a file called `~/.bashrc`, create one with the two lines above.

Then run:
```
source ~/.bashrc
```

<div class="alert alert-warning">
If you use Derecho for purposes other than this class, don't add these lines blindly to your
<code>~/.bashrc</code> — talk to us first so we can figure out the best way to set your project
number for this tutorial without disrupting your other work.
</div>


## Exercise 0: Prepare our space

Let's prepare our space by creating two directories in your home directory:
```
mkdir ~/code
mkdir ~/cases
```
The other directories we will need (build/run and archive directories) are created
automatically by CESM, or already exist for you.

<div class="alert alert-info">
<strong>Note:</strong> On Derecho, <code>~</code> is a shortcut for your home directory,
<code>/glade/u/home/$USER</code>.
</div>
