# 8.2 Exercise 6: Hybrid run — initializing a run from a previous run

<div class="alert alert-info">

<strong>Exercise 6</strong>

Create a case `b1850_hybrid` using the `B1850` compset at `f19_g17` resolution. Set the run as
a hybrid run branching from the run `b.e21.B1850.f09_g17.CMIP6-picontrol.001`. The restart
files for the reference run are in:
```
/glade/campaign/cesm/cesmdata/inputdata/cesm2_init/b.e21.B1850.f09_g17.CMIP6-picontrol.001/0501-01-01
```

</div>


```bash
# go into the scripts directory of the CESM code and create a new case
cd ~/code/release-cesm2.1.5/cime/scripts
./create_newcase --case ~/cases/b1850_hybrid --compset B1850 --res f19_g17

# go into the case directory and run case.setup
cd ~/cases/b1850_hybrid
./case.setup

# set up a hybrid run
./xmlchange RUN_REFCASE=b.e21.B1850.f09_g17.CMIP6-piControl.001
./xmlchange RUN_REFDATE=0501-01-01
./xmlchange RUN_STARTDATE=0001-01-01
./xmlchange GET_REFCASE=FALSE
./xmlchange RUN_TYPE=hybrid

# copy the reference case's initial conditions into the run directory
cp /glade/campaign/cesm/cesmdata/inputdata/cesm2_init/b.e21.B1850.f09_g17.CMIP6-piControl.001/0501-01-01/* \
   /glade/derecho/scratch/$USER/b1850_hybrid/run

# build and submit
qcmd -- ./case.build
./case.submit
```
