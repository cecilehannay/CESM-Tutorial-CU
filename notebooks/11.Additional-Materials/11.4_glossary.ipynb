# Glossary

A quick reference for terms used throughout this tutorial. If you forget what something means,
look it up here rather than re-reading a whole section.

- **archive directory** — where CESM moves your output (history files) and restart files after
  a run completes. See [Understanding the five directories](../02.CESM-structure/2.1_navigating_maze.ipynb).
- **branch run** — a run type that starts from another run's restart files but must stay
  bit-for-bit identical to it (no namelist or source changes). See
  [Run types](../08.Runtype/8.1_overview.ipynb).
- **case** — the name you give one CESM experiment/simulation. Represented by `$CASE` in paths.
- **case directory** (`$CASEROOT`) — contains the scripts and settings for one case:
  `case.setup`, `case.build`, `case.submit`, the `env_*.xml` files, the `user_nl_*` files. See
  [Understanding the five directories](../02.CESM-structure/2.1_navigating_maze.ipynb).
- **CaseStatus** — a file in the case directory that logs a pass/fail line for every step
  (`case.setup`, `case.build`, `case.run`, `st_archive`). See
  [The 4 commands to run CESM on Derecho](../04.CESM-Workflow-Quickstart/4.1_four_commands.ipynb).
- **compset** (component set) — which model components are active, data-driven, or stubbed
  out, e.g. `B1850`. See [More info about compsets and grids](../04.CESM-Workflow-Quickstart/4.2_compsets_and_grids.ipynb).
- **CONTINUE_RUN** — an `env_run.xml` variable: `FALSE` for the first segment of a run, `TRUE`
  to continue/extend it. See [The xml files and run length](../05.Run-length/5.1_xml_files.ipynb).
- **hybrid run** — a run type that starts from another run's restart files but *can* change
  namelists, source code, and the start date. See [Run types](../08.Runtype/8.1_overview.ipynb).
- **history file** — a CESM output file (e.g. `case01.cam.h0.0001-01.nc`). Lives in the run
  directory while the model is running, then moves to the archive directory.
- **inputdata directory** — the large, shared directory of input files (grids, forcing data,
  initial conditions) that CESM reads from. Already exists on Derecho — you never create it.
- **log file** — a file recording what happened while a component ran (e.g. `cesm.log.*`,
  `atm.log.*`). The first place to look when a run crashes. See
  [My run crashed — where do I find info?](../09.Troubleshooting/9.1_log_files_and_crashes.ipynb).
- **namelist** — a set of settings a component reads at runtime (e.g. `co2vmr`). You edit
  these through `user_nl_*` files. See [CESM namelists](../07.Namelist/7.1_namelist_overview.ipynb).
- **project code** — the allocation charged for your compute time on Derecho (e.g.
  `UCUB0184`). See [Project code issues](../09.Troubleshooting/9.2_project_code_issues.ipynb).
- **resolution / grid** — how finely the model divides up the Earth's surface, set with
  `--res` (e.g. `f19_g17`). See [More info about compsets and grids](../04.CESM-Workflow-Quickstart/4.2_compsets_and_grids.ipynb).
- **restart file** — a file that saves the model's exact state so a run can be continued or
  branched from later (e.g. `case01.cam.r.*`). See [Restart files](../05.Run-length/5.2_restart_files.ipynb).
- **RESUBMIT** — an `env_run.xml` variable: how many times a case should automatically
  resubmit itself once a segment finishes. See [The xml files and run length](../05.Run-length/5.1_xml_files.ipynb).
- **run directory** (`$RUNDIR`) — where CESM actually builds and executes the model, and
  where history/log files live while a run is in progress. See
  [Understanding the five directories](../02.CESM-structure/2.1_navigating_maze.ipynb).
- **startup run** — the ordinary run type: a cold start (no restart files) for the first
  segment, or a continuation of the *same* run (`CONTINUE_RUN=TRUE`) with nothing else allowed
  to change. See [Run types](../08.Runtype/8.1_overview.ipynb).
- **STOP_N / STOP_OPTION** — `env_run.xml` variables that set how long a run segment lasts
  (e.g. `STOP_N=5`, `STOP_OPTION=ndays`). See [The xml files and run length](../05.Run-length/5.1_xml_files.ipynb).
- **user_nl_\*** — files in the case directory where you make namelist changes (e.g.
  `user_nl_cam`). See [CESM namelists](../07.Namelist/7.1_namelist_overview.ipynb).
- **`$USER` / `$CASE` / `~`** — shell shortcuts used throughout this tutorial: `$USER` is your
  Derecho username, `$CASE` is your case name, and `~` is your home directory
  (`/glade/u/home/$USER`).
- **xmlquery / xmlchange** — the commands to look up (`xmlquery`) or change (`xmlchange`) a
  variable in a case's `env_*.xml` files. See [The xml files and run length](../05.Run-length/5.1_xml_files.ipynb).
