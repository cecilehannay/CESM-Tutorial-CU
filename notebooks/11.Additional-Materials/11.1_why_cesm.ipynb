# 11.1 Why do you want to use CESM?

![CESM overview](../../images/cesm_overview_illustration.png)

The Community Earth System Model (CESM) is a state-of-the-art, fully-coupled Earth system
model that is freely available to the research community, used worldwide for climate
research and education.


## CESM simulations are freely available

You don't always need to run CESM yourself — a huge number of simulations have already been
run and are freely available for download:

- [NCAR Climate Data Gateway](https://www.earthsystemgrid.org/) — many simulations available
  for download, including Large Ensembles and long pre-industrial control runs.

![Earth System Grid gateway](../../images/earthsystemgrid_gateway.png)

### More CESM simulations freely available

- Climate variability and change simulations
- Subseasonal to decadal predictions
- Paleo Working Group simulations
