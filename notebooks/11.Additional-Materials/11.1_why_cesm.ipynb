# Why do you want to use CESM?

![CESM overview](../../images/cesm_overview_illustration.png)

The Community Earth System Model (CESM) is a state-of-the-art, fully coupled Earth system
model that is freely available to the research community and used worldwide for climate
research and education.

## CESM simulations are freely available

You do not always need to run CESM yourself. A huge number of simulations have already been
run and are freely available for download from the
[NCAR Climate Data Gateway](https://www.earthsystemgrid.org/), including large ensembles and
long pre-industrial control runs.

![Earth System Grid gateway](../../images/earthsystemgrid_gateway.png)

Other freely available simulations include:

- climate variability and change simulations,
- subseasonal to decadal predictions,
- Paleo Working Group simulations.

![More CESM simulations available](../../images/cesm_simulations_gateway.png)
