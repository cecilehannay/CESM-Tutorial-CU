## Increase the orographic height over the western US

https://ncar.github.io/CESM-Tutorial/notebooks/challenge/cam/exercise_4.html

This exercise edits a boundary-condition input file and uses `SourceMods` rather than a simple
namelist change. It is a good next step once you are comfortable with `user_nl_*` files.

## Increase the prescribed SST by 2 K in the tropical central Pacific

https://ncar.github.io/CESM-Tutorial/notebooks/challenge/cam/exercise_5.html

:::{note}
The prescribed sea surface temperature (SST) is not a namelist setting. It is changed with
`xmlchange SSTICE_DATA_FILENAME=...`, which points the model at a different SST input file.
:::
