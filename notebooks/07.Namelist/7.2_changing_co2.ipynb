# Example: changing the CO2 concentration

CO2 is set differently depending on the compset:

- **B1850** cycles over a single pre-industrial year, so CO2 is a single value that applies
  everywhere and all year. It is set with the namelist variable `co2vmr`.
- **BHIST** follows the 20th century, so CO2 varies from year to year. It is read from a
  time series file set with the namelist variable `flbc_file`.

To change a namelist variable, add it to the `user_nl_*` file of the right component in your
case directory. CO2 belongs to the atmosphere model, so edit `user_nl_cam`:

```fortran
&cam_inparm
co2vmr = 800.0e-6
/
```

`co2vmr` is a volume mixing ratio: the fraction of air molecules that are CO2. A value of
`800.0e-6` means 800 parts per million (ppm).

:::{important}
After editing a `user_nl_*` file, run `./preview_namelists` to regenerate the namelists (or
simply build and submit again, which regenerates them for you). Otherwise the model keeps using
the old values.
:::
