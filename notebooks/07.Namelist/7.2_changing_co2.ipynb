# 7.2 Example: changing namelist variables — setting CO2 concentration

CO2 is set differently depending on the compset:

- **B1850** → cycles over a single year → CO2 is a single global/annual value → set with the
  namelist variable `co2vmr`.
- **BHIST** → historical 20th century run → CO2 is set as a 20th-century timeseries (varies
  year to year) → set with the namelist variable `flbc_file`.

To change a namelist variable, add it to the relevant `user_nl_*` file in your case
directory. For example, to change `co2vmr` for CAM, edit `user_nl_cam`:
```
&cam_inparm
co2vmr = 800.0e-6
/
```

<div class="alert alert-info">
After editing a <code>user_nl_*</code> file, rerun <code>./case.build</code> and
<code>./case.submit</code> (or, at minimum, <code>./preview_namelists</code>) so the change is
picked up.
</div>
