# CESM namelists

Each component has its own namelist. You can look at your namelists in the run directory:
they are the files ending in `_in`.

The namelist contains a set of variables used at runtime, for example:
- `co2vmr`
- `dust_emis_fact`
- `bnd_topo`

The main namelist files you'll encounter in `$RUNDIR` are: `atm_in`, `cism_in`, `drv_flds_in`,
`drv_in`, `ice_in`, `lnd_in`, `mosart_in`, `pop_in`, `wav_in`.


## Overview of CESM directories: namelist files

The path from case directory to namelist files:

```
~/cases/case01                              (CASE directory)
    (2) case.setup
    (3) case.build → (4) case.submit
        user_nl_cam, user_nl_cice, user_nl_cism, user_nl_clm,
        user_nl_cpl, user_nl_mosart, user_nl_pop, user_nl_ww3
        SourceMods/
        CaseDocs/
            atm_in, cism_in, drv_flds_in, drv_in, ice_in,
            lnd_in, mosart_in, pop_in, wav_in, ...

/glade/derecho/scratch/$USER/case01           (Build/Run directory, $RUNDIR)
    bld/                                       ($EXEROOT)
    run/
        atm_in, cism_in, drv_flds_in, drv_in, ice_in,
        lnd_in, mosart_in, pop_in, wav_in, ...
```

You edit namelist variables through the `user_nl_*` files in your **case directory**. CESM
takes what you put there and generates the actual namelist files (`atm_in`, `lnd_in`, ...) in
`CaseDocs/` and in the run directory.

The script `preview_namelists` (called automatically by `case.build` and `case.submit`)
generates the namelist files from your `user_nl_*` edits.
