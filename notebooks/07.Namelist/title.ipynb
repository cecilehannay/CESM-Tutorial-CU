# Namelist modifications

Many model settings can be changed without touching the source code or rebuilding the
model: the CO2 concentration, how often output is written, which input files are read, and
many more. These settings live in **namelists**. In this chapter, you will learn what a
namelist is, where CESM keeps them, how to change one safely, and where to look up the
hundreds of available settings. You will then double the CO2 concentration in a new case.
