# Where to find information about namelists

There are hundreds of namelist variables, and nobody remembers them all. The
[CESM2 webpage](http://www.cesm.ucar.edu/models/cesm2/) is the place to look them up. Under
**Components Configuration Settings**, open **Component Fortran Namelist settings**.

![CESM2 component configuration settings](../../images/namelist_docs_settings.png)

## Namelist definitions for every component

Each component (atmosphere, land, ocean, ...) has its own page of namelist definitions.

![Namelist definitions](../../images/namelist_definitions.png)

## Browsing variables

You can browse the variable names and see the details of each one: its default value, the
values it accepts, and which component uses it. This is how you find out, for example, that
CO2 is called `co2vmr` and lives in `user_nl_cam`.

![Browse namelist variables](../../images/namelist_browse_variables.png)
