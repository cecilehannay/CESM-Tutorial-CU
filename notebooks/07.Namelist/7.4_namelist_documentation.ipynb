# 7.4 Where to find info about namelists

The [CESM2 webpage](http://www.cesm.ucar.edu/models/cesm2/) is the place to look up
namelist variables. Under **Components Configuration Settings**, you can find information
about namelist variables in **Component Fortran Namelist settings**.

![CESM2 component configuration settings](../../images/namelist_docs_settings.png)


## Namelist definitions for every component

![Namelist definitions](../../images/namelist_definitions.png)


## Browsing variables

You can browse variable names and see details about each one (default value, valid values,
which component uses it, ...).
