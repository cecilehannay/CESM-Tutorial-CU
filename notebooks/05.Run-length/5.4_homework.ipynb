# Homework: a two-month run in two submissions

This homework puts together everything from this section: creating a case, setting its length, and making it resubmit itself automatically.

:::{admonition} Homework
:class: exercise

Create a case called `b1850_hw` with the `B1850` compset at `f19_g17` resolution, in the
`cases` subdirectory of your home directory. Make a 2-month run using two consecutive 1-month
submissions (i.e., you want the run to automatically resubmit itself at the end of the first
month).

Explicitly specify the values you used for the variables `STOP_N`, `STOP_OPTION`,
`RESUBMIT`, `CONTINUE_RUN`.

Send the list of commands you used to both Aneesh and Cecile on Slack!
:::

:::{admonition} Hints
:class: hint dropdown

- You need the four commands from [The 4 commands to run CESM](../04.CESM-Workflow-Quickstart/4.1_four_commands.ipynb), plus a few `xmlchange` commands in between `case.setup` and `case.submit`.
- The four settings to think about are `STOP_N`, `STOP_OPTION`, `RESUBMIT`, and `CONTINUE_RUN`. The [xml files](5.1_xml_files.ipynb) page explains what each one does and shows a worked example.
- This is a new case, so ask yourself whether it should start from scratch or continue an existing run.
:::

:::{warning}
**Before you hit "case.submit"** — estimate the resources needed to complete the
run you are requesting. Be careful when setting `STOP_N` and/or
`RESUBMIT` to a large value. Don't use the premium queue outside of class. If in
doubt, please ask — our project code has finite shared resources.
:::
