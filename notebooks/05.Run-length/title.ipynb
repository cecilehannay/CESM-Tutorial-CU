# Change the length of a run

By default, a brand-new CESM case runs for only 5 simulated days. That is long enough to check that everything works, but far too short for a real experiment.

In this section, you will learn where that setting lives, how to change it, and how to continue a run that has already finished so that it keeps going from where it stopped. The homework at the end puts all of these pieces together in a two-month run.
