# The xml files and run length

## Overview of CESM directories: the xml files

The case directory contains a set of `env_*.xml` files. These contain variables used by the
CESM scripts. Some of these variables can be changed by the user:

- `env_case.xml`: set by `create_newcase`; **cannot** be modified afterwards.
- `env_mach_pes.xml`: specifies the layout of components (how many processors, etc.).
- `env_build.xml`: specifies build information.
- `env_run.xml`: sets run-time information (length of run, number of submissions, ...).

The user interacts with `env_run.xml` most frequently.


## Useful variables in `env_run.xml`

The length of the run is set with the variables `STOP_N` and `STOP_OPTION`.

By default, the model is set to run for 5 days:
```
STOP_N = 5
STOP_OPTION = ndays
```

You can query the value of a variable with `xmlquery`:
```bash
./xmlquery STOP_OPTION,STOP_N
```

You can change the value of a variable with `xmlchange`. For example, to increase the run
length from 5 days to 1 month:
```bash
./xmlchange STOP_OPTION=nmonths,STOP_N=1
```

## `CONTINUE_RUN` and `RESUBMIT`

Other useful variables in `env_run.xml`:

- `CONTINUE_RUN`
  - `FALSE`: this is the first segment of your run.
  - `TRUE`: you are restarting/continuing a run.
- `RESUBMIT`: if greater than 0, the case will automatically resubmit itself that many times
  once the current segment finishes.

For example, if you set:
```
STOP_N = 1
STOP_OPTION = nmonths
RESUBMIT = 3
```
the run will proceed one month at a time, and resubmit itself automatically 3 times — for a
total of 4 months.

More information about the variables in `env_run.xml`:
http://www.cesm.ucar.edu/models/cesm2/settings/current/drv_input.html
